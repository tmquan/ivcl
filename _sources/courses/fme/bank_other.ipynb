**Razavi, Example 5.1** (no transistor)

A voltage amplifier with a gain of 10 listens to a microphone and drives a speaker. The microphone is a source of $10\,\mathrm{mV}$ peak-to-peak with a series resistance of $200\,\Omega$; the speaker is an $8\,\Omega$ resistor. The amplifier has an input resistance $R_{in}$ to ground and an output resistance $R_{amp}$ in series with its ideal output voltage. (a) How much of the microphone signal appears across the amplifier input if $R_{in}=2\,\mathrm{k}\Omega$ or $500\,\Omega$? (b) What fraction of the amplifier output voltage reaches the speaker if $R_{amp}=10\,\Omega$ or $2\,\Omega$?

Given: $v_m=10\,\mathrm{mV_{pp}}$, $R_S=200\,\Omega$, $R_L=8\,\Omega$, $A_v=10$, $R_{in}=2\,\mathrm{k}\Omega$ or $500\,\Omega$, $R_{amp}=10\,\Omega$ or $2\,\Omega$

1. **Input interface.** The microphone resistance $R_S$ and the amplifier input resistance $R_{in}$ form a voltage divider: $v_{in}=\dfrac{R_{in}}{R_{in}+R_S}\,v_m$.
2. **(a) $R_{in}=2\,\mathrm{k}\Omega$.** $v_{in}=\dfrac{2000}{2000+200}v_m=0.909\,v_m$, i.e. $9.09\,\mathrm{mV_{pp}}$ of the $10\,\mathrm{mV_{pp}}$: a loss of only about 9 %.
3. **(a) $R_{in}=500\,\Omega$.** $v_{in}=\dfrac{500}{500+200}v_m=0.714\,v_m$, i.e. $7.14\,\mathrm{mV_{pp}}$: nearly 30 % is lost. A large input resistance is desirable.
4. **Output interface.** The ideal amplifier output $v_{amp}$ sits behind $R_{amp}$ and drives the speaker $R_L$: $v_{out}=\dfrac{R_L}{R_L+R_{amp}}\,v_{amp}$.
5. **(b) $R_{amp}=10\,\Omega$.** $v_{out}=\dfrac{8}{8+10}v_{amp}=0.444\,v_{amp}$: more than half of the signal is lost.
6. **(b) $R_{amp}=2\,\Omega$.** $v_{out}=\dfrac{8}{8+2}v_{amp}=0.800\,v_{amp}$. A small output resistance is desirable.
7. **Numerical illustration.** With an ideal input ($v_{amp}=10\times10\,\mathrm{mV_{pp}}=100\,\mathrm{mV_{pp}}$) the speaker would see $44.4\,\mathrm{mV_{pp}}$ or $80.0\,\mathrm{mV_{pp}}$ respectively.

**Result.** (a) $0.909\,v_m$ for $2\,\mathrm{k}\Omega$, $0.714\,v_m$ for $500\,\Omega$; (b) $0.444\,v_{amp}$ for $10\,\Omega$, $0.800\,v_{amp}$ for $2\,\Omega$

Checked: agrees with the answer in the book.

**Razavi, Example 10.1** (BJT)

A microphone signal is amplified by a single common-emitter stage (collector resistor $R_C$, collector current $I_C$) whose supply $V_{CC}$ comes from a rectifier. The rectifier output is not perfectly constant but carries a ripple $v_r$ at twice the line frequency. The loudspeaker plays a loud steady low-frequency hum. Explain where it comes from and how strongly the ripple reaches the output.

Given: $V_{CC}$ with ripple $v_r$ (frequency $2f_\mathrm{line}$); $V_{out}=V_{CC}-R_CI_C$

1. **Output voltage of the stage.** The collector node sits at the supply minus the drop across the load: $V_{out}=V_{CC}-R_CI_C$.
2. **Bias current is set by the input.** $I_C$ is fixed by the base bias, not by $V_{CC}$ (ignoring the Early effect), so a change of $V_{CC}$ leaves $R_CI_C$ unchanged.
3. **Effect of the ripple.** If $V_{CC}$ moves by $v_r$, then $V_{out}$ moves by $v_r$ as well: $\Delta V_{out}/\Delta V_{CC}=1$. The output therefore follows the supply with unity gain.
4. **Two components at the output.** $v_{out}=A_vv_{in}+v_r$: the amplified microphone signal plus the whole ripple.
5. **Source of the hum.** The ripple is a steady component at $2f_\mathrm{line}$ (100 or 120 Hz), which the loudspeaker reproduces as hum. The cause is that $V_{out}$ is measured against ground while $V_{CC}$ is not constant; the output is not insensitive to the supply.
6. **Remedy hinted by the book.** Measure the output between this node and a second node that carries the same ripple (a duplicate stage), so the ripple cancels in the difference: this is the idea of the differential amplifier.

**Result.** Hum comes from supply ripple reaching $V_{out}$ with gain $1$: $v_{out}=A_vv_{in}+v_r$

Checked: agrees with the answer in the book.

**Razavi, Example 10.2** (no transistor)

A transformer with a center-tapped secondary delivers two signals $+v_{in}$ and $-v_{in}$ that are equal and opposite about the center tap. With the tap grounded their common-mode (average) level is 0 V. How must the transformer be connected so the two outputs have a common-mode level of $+2\,\mathrm{V}$?

Given: Required common-mode level $V_{CM}=2\,\mathrm{V}$

1. **Common-mode level.** The CM level of two outputs $V_1,V_2$ that swing equally and oppositely is their dc average, which is the dc level of the center tap.
2. **Outputs with the tap grounded.** $V_1=+v_{in}$, $V_2=-v_{in}$, so $(V_1+V_2)/2=0$.
3. **Raise the tap.** The windings only fix the voltage of each end relative to the tap. Tying the center tap to $+2\,\mathrm{V}$ shifts both ends by $2\,\mathrm{V}$: $V_1=2\,\mathrm{V}+v_{in}$, $V_2=2\,\mathrm{V}-v_{in}$.
4. **Check.** Average $=2\,\mathrm{V}$ and difference $V_1-V_2=2v_{in}$ unchanged.

**Result.** Connect the center tap to a $+2\,\mathrm{V}$ source: $V_{CM}=2\,\mathrm{V}$

Checked: agrees with the answer in the book.

**Razavi, Problem 5.1** (no transistor)

An antenna is modelled by its Thevenin equivalent: a sinusoidal source $V_0\cos\omega t$ in series with an output resistance $R_{out}$. A load resistor $R_L$ is connected across it. Find the average power delivered to $R_L$ and sketch it as a function of $R_L$.

Given: $v_s=V_0\cos\omega t$, source resistance $R_{out}$, load $R_L$

1. **Voltage divider.** The load voltage is $v_L=\dfrac{R_L}{R_{out}+R_L}V_0\cos\omega t$.
2. **Instantaneous power.** $p(t)=\dfrac{v_L^2}{R_L}=\dfrac{R_LV_0^2}{(R_{out}+R_L)^2}\cos^2\omega t$.
3. **Average over a period.** The mean of $\cos^2\omega t$ is $1/2$, so $P_{av}=\dfrac{V_0^2R_L}{2(R_{out}+R_L)^2}$.
4. **Maximum.** $\dfrac{dP_{av}}{dR_L}\propto (R_{out}+R_L)-2R_L=0$ gives $R_L=R_{out}$ and $P_{max}=\dfrac{V_0^2}{8R_{out}}$.
5. **Shape of the plot.** $P_{av}=0$ at $R_L=0$, rises linearly ($\approx V_0^2R_L/2R_{out}^2$) for $R_L\ll R_{out}$, peaks at $R_L=R_{out}$, then falls as $\approx V_0^2/2R_L$ for $R_L\gg R_{out}$. It is symmetric on a logarithmic $R_L$ axis: $P_{av}(R_L)=P_{av}(R_{out}^2/R_L)$.

**Result.** $P_{av}=\dfrac{V_0^2R_L}{2(R_{out}+R_L)^2}$, maximum $V_0^2/(8R_{out})$ at $R_L=R_{out}$

**Razavi, Problem 5.2** (no transistor)

Find the small-signal input resistance of three diode networks, all diodes forward-biased (each diode is a small-signal resistance $r_D=V_T/I_D$). (a) Input through diode $D_1$, then $R_1$ to ground. (b) Input through $D_1$ to a node X; from X, $R_1$ to ground and diode $D_2$ (anode at X) to ground. (c) Input through a diode to node X; between X and node Y, $R_1$ in parallel with diode $D_2$ (anode at X); from Y a second diode $D_1$ (anode at Y) to ground.

Given: Diodes forward-biased; $r_{D}=V_T/I_D$

1. **Diode model.** Each forward-biased diode is replaced by $r_{Dj}=V_T/I_{Dj}$; a constant voltage source is not needed for small signals.
2. **(a) Series path.** $R_{in}=r_{D1}+R_1$.
3. **(b) Series then parallel.** $R_{in}=r_{D1}+(R_1\parallel r_{D2})$.
4. **(c) Same current in the two series diodes.** The input diode and the diode to ground carry the same current, so both have $r_{D1}=V_T/I_{D1}$.
5. **(c) Total.** $R_{in}=2r_{D1}+(R_1\parallel r_{D2})$.
6. **Numerical check.** With $I_{D1}=1\,\mathrm{mA}$, $I_{D2}=0.4\,\mathrm{mA}$, $R_1=100\,\Omega$: $r_{D1}=26\,\Omega$, $r_{D2}=65\,\Omega$, (c) gives $91.4\,\Omega$ by formula and by nodal analysis.

**Result.** (a) $r_{D1}+R_1$; (b) $r_{D1}+R_1\parallel r_{D2}$; (c) $2r_{D1}+R_1\parallel r_{D2}$

**Razavi, Problem 6.2** (MOSFET)

A MOSFET is in pinch-off near the drain. The relation $I=Q_d\,v$ (current = mobile charge per unit length times carrier velocity) implies that if the current is constant, charge density and velocity change in opposite directions. How should this be interpreted at the pinch-off point, where the charge density tends to zero?

Given: $I_D=Q_d(x)\,v(x)$ constant along the channel

1. **What the equation says.** Since $I_D=Q_d v$ is the same at every $x$, a falling $Q_d$ must be compensated by a rising $v$; as $Q_d\to 0$ at pinch-off, $v=I_D/Q_d$ would have to tend to infinity.
2. **Why that cannot be literal.** Carrier velocity is limited (it saturates near $10^7\,\mathrm{cm/s}$ in silicon), so the charge density never truly reaches zero: a small but finite density of electrons moving very fast carries the same current.
3. **Where the model breaks.** The relation $Q_d=WC_{ox}(V_{GS}-V-V_{TH})$ assumes the vertical field dominates (gradual-channel approximation). Near pinch-off the lateral field becomes large, the approximation fails, and the "zero charge" is an artifact of the simple model.
4. **Beyond pinch-off.** Past the pinch-off point the electrons leave the inversion layer and are swept through the depletion region to the drain by the strong lateral field, much as carriers cross the collector junction of a bipolar transistor; the current is set by how many electrons reach the pinch-off point, which is why $I_D$ barely depends on $V_{DS}$.

**Result.** The charge density becomes very small but not zero, and the carriers move very fast (up to the saturation velocity), so $Q_d v$ stays equal to $I_D$; beyond pinch-off the high lateral field sweeps the carriers to the drain.

**Razavi, Problem 6.4** (MOSFET)

For the NMOS channel of Fig. 6.11 (source at $x=0$, drain at $x=L$, gate above threshold) with a positive drain voltage $V_D$ that keeps the device in triode, (a) sketch the electron density in the channel versus $x$; (b) sketch the local channel resistance per unit length versus $x$.

Given: $V_S=0$, $0< V_D< V_{GS}-V_{TH}$ (triode); $Q_d(x)=WC_{ox}[V_{GS}-V(x)-V_{TH}]$

1. **Channel potential.** Solving the channel equation at constant $I_D$ (Problem 6.5) gives $V(x)=V_{ov}-\sqrt{V_{ov}^2-\dfrac{2I_Dx}{\mu_nC_{ox}W}}$ with $V_{ov}=V_{GS}-V_{TH}$; $V$ rises from 0 to $V_D$, slowly at first and faster near the drain.
2. **(a) Electron density.** $n(x)=Q_d(x)/q=\dfrac{WC_{ox}}{q}\sqrt{V_{ov}^2-\dfrac{2I_Dx}{\mu_nC_{ox}W}}$ (per unit length): maximum $WC_{ox}V_{ov}/q$ at the source, falling along $x$ as a square root, to $WC_{ox}(V_{ov}-V_D)/q$ at the drain. The fall is steepest near the drain.
3. **(a) Limit.** At the edge of saturation ($V_D=V_{ov}$) the density falls to zero at $x=L$ with a vertical tangent: pinch-off.
4. **(b) Local resistance.** A slice $dx$ has resistance $dR=\dfrac{dx}{\mu_nQ_d(x)}$, so $\dfrac{dR}{dx}=\dfrac{1}{\mu_nWC_{ox}[V_{ov}-V(x)]}$: minimum $1/(\mu_nWC_{ox}V_{ov})$ at the source, rising along the channel (as one over a square root) to $1/(\mu_nWC_{ox}(V_{ov}-V_D))$ at the drain, and tending to infinity there at pinch-off.
5. **Consistency.** $dV/dx=I_D\,dR/dx$: the potential climbs fastest where the local resistance is highest, near the drain.

**Result.** (a) Electron density largest at the source, decreasing as $\sqrt{V_{ov}^2-2I_Dx/(\mu_nC_{ox}W)}$ to $WC_{ox}(V_{ov}-V_D)/q$ at the drain; (b) resistance per unit length $1/(\mu_nWC_{ox}[V_{ov}-V(x)])$, smallest at the source, rising toward the drain

**Razavi, Problem 6.5** (MOSFET)

With the drain current $I_D$ constant along the channel, solve the channel equation $I_D=WC_{ox}[V_{GS}-V(x)-V_{TH}]\,\mu_n\,dV/dx$ (Eq. 6.7) for the channel potential $V(x)$, with $V(0)=0$. Plot $V(x)$ and $dV/dx$ versus $x$ for different values of $W$ or $V_{TH}$.

Given: $I_D=\mu_nC_{ox}W[V_{GS}-V(x)-V_{TH}]\,dV/dx$, $V(0)=0$

1. **Separate variables.** $I_D\,dx=\mu_nC_{ox}W\,(V_{ov}-V)\,dV$ with $V_{ov}=V_{GS}-V_{TH}$.
2. **Integrate from the source.** $I_Dx=\mu_nC_{ox}W\left(V_{ov}V-\tfrac12V^2\right)$.
3. **Solve the quadratic.** $V(x)=V_{ov}-\sqrt{V_{ov}^2-\dfrac{2I_Dx}{\mu_nC_{ox}W}}$ (the root with $V(0)=0$).
4. **Slope.** $\dfrac{dV}{dx}=\dfrac{I_D}{\mu_nC_{ox}W\sqrt{V_{ov}^2-2I_Dx/(\mu_nC_{ox}W)}}=\dfrac{I_D}{\mu_nC_{ox}W\,[V_{ov}-V(x)]}$.
5. **Shapes.** $V(x)$ starts at 0 with slope $I_D/(\mu_nC_{ox}WV_{ov})$ and curves upward (concave up); $dV/dx$ grows along $x$ and diverges where $V\to V_{ov}$ (pinch-off).
6. **Effect of $W$.** For a given $I_D$, a larger $W$ lowers $2I_Dx/(\mu_nC_{ox}W)$: $V(x)$ rises more slowly and $dV/dx$ is smaller everywhere (a wider channel needs less field to carry the same current).
7. **Effect of $V_{TH}$.** A larger $V_{TH}$ lowers $V_{ov}$: the curve starts steeper ($dV/dx$ at $x=0$ is $I_D/(\mu_nC_{ox}WV_{ov})$) and approaches the pinch-off limit $V=V_{ov}$ sooner, so $V(x)$ rises faster and $dV/dx$ is larger.

**Result.** $V(x)=V_{GS}-V_{TH}-\sqrt{(V_{GS}-V_{TH})^2-\dfrac{2I_Dx}{\mu_nC_{ox}W}}$, $\dfrac{dV}{dx}=\dfrac{I_D}{\mu_nC_{ox}W[V_{GS}-V_{TH}-V(x)]}$

**Razavi, Problem 6.12** (MOSFET)

For a MOSFET used as a resistor (deep triode), define the intrinsic time constant $\tau=R_{on}C_{GS}$ with $C_{GS}=WLC_{ox}$. Find an expression for $\tau$ and say what the designer must do to minimize it.

Given: $R_{on}=\dfrac{1}{\mu_nC_{ox}(W/L)(V_{GS}-V_{TH})}$, $C_{GS}=WLC_{ox}$

1. **Multiply.** $\tau=\dfrac{WLC_{ox}}{\mu_nC_{ox}(W/L)(V_{GS}-V_{TH})}$.
2. **Simplify.** $\tau=\dfrac{L^2}{\mu_n(V_{GS}-V_{TH})}$: independent of $W$ and of $C_{ox}$.
3. **Minimize.** Use the shortest channel length available ($\tau\propto L^2$) and the largest overdrive $V_{GS}-V_{TH}$ (drive the gate as high as possible); higher carrier mobility also helps (an NMOS rather than a PMOS). Changing $W$ does not help: it lowers $R_{on}$ and raises $C_{GS}$ by the same factor.

**Result.** $\tau=\dfrac{L^2}{\mu_n(V_{GS}-V_{TH})}$; minimize with small $L$ and large $V_{GS}-V_{TH}$

2. **Open-circuit gain.** With $R_L$ open: $i_f=g_mv_i+v_o/R_2$ and $v_i=i_fR_f+v_o$, so $v_i(1-g_mR_f)=v_o(1+R_f/R_2)$ and $A_{vo}=\dfrac{1-g_mR_f}{1+R_f/R_2}$.
3. **Output resistance.** With $v_i=0$ the source is off: $R_o=R_2\parallel R_f$.
4. **Numbers: $R_{in}$.** $R_2\parallel R_L=0.0909\,\mathrm{k\Omega}$; $R_{in}'=\dfrac{1000.09}{1+9.09}=99.1\,\mathrm{k\Omega}$; $R_{in}=100\parallel99.1=49.8\,\mathrm{k\Omega}$.
5. **Numbers: $A_{vo}$, $R_o$.** $A_{vo}=\dfrac{1-10^5}{1+10^4}=-10.0\,\mathrm{V/V}$; $R_o=0.1\parallel1000=100\,\Omega$.
6. **Overall gain.** $G_v=\dfrac{49.8}{49.8+100}(-10)\dfrac{1}{1+0.1}=-3.02\,\mathrm{V/V}$.
7. **Without $R_f$.** $R_{in}=100\,\mathrm{k\Omega}$, $A_{vo}=-g_mR_2=-10$, $R_o=100\,\Omega$, $G_v=\dfrac{100}{200}(-10)\dfrac1{1.1}=-4.55\,\mathrm{V/V}$: only $R_{in}$ is changed noticeably.

**Result.** $R_{in}=49.8\,\mathrm{k\Omega}$, $A_{vo}=-10\,\mathrm{V/V}$, $R_o=100\,\Omega$, $G_v=-3.02\,\mathrm{V/V}$ (without $R_f$: $-4.55$)

Checked: agrees with the answer in the solutions manual. Note: Circuit read from the manual figure (Rf from input to output node, gm*vi and R2||RL at the output). Rsig=100k comes from the manual gain arithmetic. The nodal check at the end reproduces Rin.

**Sedra/Smith, Problem 6.137** (BJT)

A common-emitter amplifier biased at $I_C=1\,\mathrm{mA}$ has midband gain $100\,\mathrm{V/V}$, lower 3-dB frequency $f_L=100\,\mathrm{Hz}$ and upper 3-dB frequency $f_H=500\,\mathrm{kHz}$. A small emitter resistance $R_e$ is added to raise $f_H$ to $2\,\mathrm{MHz}$. Assume the usual single-pole feedback behavior: gain is divided by $1+g_mR_e$ while both band edges are scaled by this factor (gain falls, $f_H$ rises, $f_L$ falls). Find $R_e$, the new $f_L$ and the new midband gain ($V_T=25\,\mathrm{mV}$).

Given: $I_C=1\,\mathrm{mA}$, $\lvert A_M\rvert=100$, $f_L=100\,\mathrm{Hz}$, $f_H=500\,\mathrm{kHz}\to2\,\mathrm{MHz}$

1. **Required factor.** $\dfrac{f_H^{\prime}}{f_H}=\dfrac{2\,\mathrm{MHz}}{500\,\mathrm{kHz}}=4=1+g_mR_e$.
2. **Transconductance.** $g_m=\dfrac{I_C}{V_T}=\dfrac{1}{0.025}=40\,\mathrm{mA/V}$.
3. **Emitter resistance.** $R_e=\dfrac{4-1}{g_m}=\dfrac{3}{0.04\,\mathrm{S}}=75\,\Omega$.
4. **New lower cutoff.** $f_L^{\prime}=\dfrac{100\,\mathrm{Hz}}{1+g_mR_e}=\dfrac{100}{4}=25\,\mathrm{Hz}$.
5. **New midband gain.** $\lvert A_M^{\prime}\rvert=\dfrac{100}{4}=25\,\mathrm{V/V}$ The product $A_Mf_H=5\times10^7\,\mathrm{Hz}$ is the same before ($100\times500\,\mathrm{kHz}$) and after ($25\times2\,\mathrm{MHz}$).

**Result.** $R_e=75\,\Omega$, $f_L=25\,\mathrm{Hz}$, $\lvert A_M\rvert=25\,\mathrm{V/V}$

Checked: agrees with the answer in the solutions manual. *Needs review.* Note: The manual gives only the solution; the scaling of $f_L$ by $1/(1+g_mR_e)$ is its assumption (strictly, the low-frequency pole depends on which capacitor dominates), and $A_M=100$, $f_L=100$ Hz are read from its numbers. Gain falls by the same factor 4, so $A_Mf_H$ is constant.
