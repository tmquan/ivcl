**Problem 1**

| | BJT |
|---|---|
| **Problem** | The circuit of BJT Problem 72 of the page One Transistor is changed: the collector is now joined to the $1.8\,\mathrm{V}$ supply through $R_C=100\,\Omega$, and the output is the collector voltage $V_{out}$. Base drive as before: $800\,\mathrm{mV}$ battery plus $v_1$; $I_S=3\times10^{-16}\,\mathrm{A}$, $\beta=100$. (a) Verify that $Q_1$ is in the active mode. (b) Find the change of $V_{out}$ for $v_1=1\,\mathrm{mV}$. |
| **Given** | $I_S=3\times10^{-16}\,\mathrm{A}$, $V_{BE}=0.8\,\mathrm{V}$, $V_{CC}=1.8\,\mathrm{V}$, $R_C=100\,\Omega$, $v_1=1\,\mathrm{mV}$ |
| **Answer 1** | **(a) Bias current.** $I_C$ $=I_S\exp(V_{BE}/V_T)$ $=6.92\,\mathrm{mA}$ (from BJT Problem 72 of the page One Transistor). |
| **Answer 2** | **(a) Drop on $R_C$.** $R_CI_C$ $=100\,\Omega\times6.92\,\mathrm{mA}$ $=0.692\,\mathrm{V}$. |
| **Answer 3** | **(a) Output bias.** $V_{out}$ $=V_{CC}-R_CI_C$ $=1.8-0.692$ $=1.108\,\mathrm{V}$. |
| **Answer 4** | **(a) Region.** $V_C$ $=1.108\,\mathrm{V}>V_B$ $=0.8\,\mathrm{V}$, so the B-C junction is reverse-biased: active mode. |
| **Answer 5** | **(b) Change in collector current.** From BJT Problem 72 of the page One Transistor, $\Delta I_C$ $=g_mv_1$ $=0.266\,\mathrm{mA}$ ($g_m=0.266\,\mathrm{S}$). |
| **Answer 6** | **(b) Change in output.** $\Delta V_{out}$ $=-R_C\,\Delta I_C$ $=-100\,\Omega\times0.266\,\mathrm{mA}$ $=-26.6\,\mathrm{mV}$: the output falls when the input rises. |
| **Answer 7** | **Gain.** $\lvert A_v\rvert$ $=g_mR_C$ $=0.266\,\mathrm{S}\times100\,\Omega$ $=26.6$ (negative sign). |
| **Result** | $V_{out}=1.108\,\mathrm{V}$ (active); $\Delta V_{out}=-26.6\,\mathrm{mV}$ for $v_1=1\,\mathrm{mV}$, gain $-26.6$ |

**Problem 2**

| | BJT |
|---|---|
| **Problem** | In the stage of Problem 1 ($800\,\mathrm{mV}$ bias at the base, $I_S=3\times10^{-16}\,\mathrm{A}$, so $I_C=6.92\,\mathrm{mA}$ and $g_m=1/(3.76\,\Omega)$) the collector resistor is raised to $R_C=200\,\Omega$ and the supply to $V_{CC}=3.6\,\mathrm{V}$. Verify that the transistor stays in the active mode and find the voltage gain. |
| **Given** | $R_C=200\,\Omega$, $V_{CC}=3.6\,\mathrm{V}$, $V_{BE}=0.8\,\mathrm{V}$, $I_C=6.92\,\mathrm{mA}$, $g_m=0.266\,\mathrm{S}$ |
| **Answer 1** | **Bias current.** $I_C$ depends only on $V_{BE}$, so it is still $6.92\,\mathrm{mA}$. |
| **Answer 2** | **Drop on $R_C$.** $R_CI_C$ $=200\,\Omega\times6.92\,\mathrm{mA}$ $=1.384\,\mathrm{V}$. |
| **Answer 3** | **Collector voltage.** $V_C$ $=3.6-1.384$ $=2.216\,\mathrm{V}>V_B$ $=0.8\,\mathrm{V}$: active mode. |
| **Answer 4** | **Why the supply had to rise.** With $V_{CC}=1.8\,\mathrm{V}$ one would get $V_C$ $=1.8-1.384$ $=0.416\,\mathrm{V}<0.8\,\mathrm{V}$: the B-C junction would be forward-biased by about 0.38 V and the transistor would leave the active mode. |
| **Answer 5** | **Small-signal model.** $V_{CC}$ is an ac ground, $v_\pi=v_1$, and the controlled current $g_mv_\pi$ flows from the output node through $R_C$ to ground: $v_{out}=-g_mv_\pi R_C=-g_mv_1R_C$. |
| **Answer 6** | **Gain.** $\dfrac{v_{out}}{v_1}$ $=-g_mR_C$ $=-\dfrac{200\,\Omega}{3.76\,\Omega}$ $=-53.2$: doubling $R_C$ doubles the gain of Problem 1. |
| **Result** | $V_C=2.216\,\mathrm{V}$ (active); gain $=-g_mR_C=-53.2$ |

**Problem 3**

| | BJT |
|---|---|
| **Problem** | A common-emitter stage with resistive collector load (core, Early effect neglected) must provide an input impedance $R_{in}$ and an output impedance $R_{out}$. What is its voltage gain, in terms of $R_{in}$, $R_{out}$ and $\beta$? |
| **Given** | Symbolic: $R_{in}$, $R_{out}$, $\beta$. Numerical check: $R_{in}=2.6\,\mathrm{k}\Omega$, $R_{out}=1\,\mathrm{k}\Omega$, $\beta=100$ |
| **Answer 1** | **Impedances of the CE core.** $R_{in}=r_\pi=\beta/g_m$ and $R_{out}=R_C$. |
| **Answer 2** | **Gain.** $A_v=-g_mR_C$. |
| **Answer 3** | **Eliminate $g_m$ and $R_C$.** $g_m=\beta/R_{in}$ and $R_C=R_{out}$, therefore $A_v$ $=-\dfrac{\beta}{R_{in}}R_{out}$ $=-\beta\dfrac{R_{out}}{R_{in}}$. |
| **Answer 4** | **Consequence.** Once $R_{in}$ and $R_{out}$ are specified the gain is fixed: the specifications are coupled. |
| **Answer 5** | **Numerical check.** $g_m$ $=\dfrac{100}{2.6\,\mathrm{k}\Omega}$ $=38.5\,\mathrm{mS}$ (i.e. $I_C=1\,\mathrm{mA}$), $-g_mR_C=-38.5$ and $-\beta R_{out}/R_{in}=-38.5$. |
| **Result** | $A_v$ $=-\beta\,\dfrac{R_{out}}{R_{in}}$ |

**Problem 4**

| | BJT |
|---|---|
| **Problem** | A common-emitter stage (emitter grounded, collector resistor $R_C=1\,\mathrm{k}\Omega$ to the supply, input at the base) is biased at $I_C=1\,\mathrm{mA}$. Given $\beta=100$ and Early voltage $V_A=10\,\mathrm{V}$, find the small-signal voltage gain and the input and output impedances. Use $V_T=26\,\mathrm{mV}$. |
| **Given** | $I_C=1\,\mathrm{mA}$, $R_C=1\,\mathrm{k}\Omega$, $\beta=100$, $V_A=10\,\mathrm{V}$, $V_T=26\,\mathrm{mV}$ |
| **Answer 1** | **Small-signal parameters.** $g_m$ $=\dfrac{I_C}{V_T}$ $=\dfrac1{26\,\Omega}$, $r_O$ $=\dfrac{V_A}{I_C}$ $=10\,\mathrm{k}\Omega$, $r_\pi$ $=\dfrac{\beta}{g_m}$ $=2.6\,\mathrm{k}\Omega$. |
| **Answer 2** | **Load seen by the collector.** $r_O$ is in parallel with $R_C$: $R_C\parallel r_O$ $=1\,\mathrm{k}\Omega\parallel10\,\mathrm{k}\Omega$ $=909\,\Omega$. |
| **Answer 3** | **Gain.** $A_v$ $=-g_m(R_C\parallel r_O)$ $=-\dfrac{909\,\Omega}{26\,\Omega}$ $=-35.0$. (Without the Early effect: $-38.5$.) |
| **Answer 4** | **Input impedance.** $R_{in}$ $=r_\pi$ $=2.60\,\mathrm{k}\Omega$. |
| **Answer 5** | **Output impedance.** $R_{out}$ $=R_C\parallel r_O$ $=0.909\,\mathrm{k}\Omega$. |
| **Result** | $A_v\approx-35.0$ ($-38.5$ if $V_A=\infty$), $R_{in}=2.60\,\mathrm{k}\Omega$, $R_{out}=0.909\,\mathrm{k}\Omega$ |

**Problem 5**

| | BJT |
|---|---|
| **Problem** | A common-emitter stage with grounded emitter and $R_C=1\,\mathrm{k}\Omega$ uses a hypothetical transistor that obeys $I_C=I_S\exp\frac{V_{BE}}{2V_T}$ and has no Early effect. Find the voltage gain at a bias current of $1\,\mathrm{mA}$. |
| **Given** | $I_C=I_S\exp(V_{BE}/2V_T)$, $I_C=1\,\mathrm{mA}$, $R_C=1\,\mathrm{k}\Omega$, $V_A=\infty$, $V_T=26\,\mathrm{mV}$ |
| **Answer 1** | **Transconductance.** $g_m$ $=\dfrac{\partial I_C}{\partial V_{BE}}$ $=\dfrac{I_S}{2V_T}\exp\dfrac{V_{BE}}{2V_T}$ $=\dfrac{I_C}{2V_T}$. |
| **Answer 2** | **Value.** $g_m$ $=1\,\mathrm{mA}/52\,\mathrm{mV}$ $=19.2\,\mathrm{mS}$. |
| **Answer 3** | **Gain.** $A_v=-g_mR_C=-19.2$, half of what a normal bipolar transistor gives at the same current. |
| **Result** | $A_v$ $=-g_mR_C$ $=-\dfrac{I_CR_C}{2V_T}$ $=-19.2$ |

**Problem 6**

| | BJT |
|---|---|
| **Problem** | A common-emitter stage: npn $Q_1$ with grounded emitter, collector tied to $V_{CC}$ through $R_C$, output $V_{out}$ at the collector, base held at a fixed bias (the microphone signal is coupled in through a capacitor). The supply $V_{CC}$ carries ripple. Treating $V_{CC}$ as a small-signal input, find the small-signal gain $v_{out}/v_{cc}$ with a finite Early voltage $V_A$. |
| **Given** | $R_C$, $r_O=V_A/I_C$ finite, base bias and $v_{in}$ set to zero |
| **Answer 1** | **Small-signal circuit.** With the base at ac ground, $v_{BE}=0$, so the dependent source $g_mv_{BE}$ is zero; only $r_O$ (collector to ground) remains. |
| **Answer 2** | **Divider.** $v_{out}$ is the supply ripple divided between $R_C$ (to $V_{CC}$) and $r_O$ (to ground). |
| **Answer 3** | **Gain.** $\dfrac{v_{out}}{v_{cc}}$ $=\dfrac{r_O}{r_O+R_C}$ |
| **Answer 4** | **Limit.** For $V_A\to\infty$, $r_O\to\infty$ and the gain tends to $1$: the output follows the ripple entirely. |
| **Answer 5** | **Numeric check.** For $R_C=1\,\mathrm{k}\Omega$, $r_O=10\,\mathrm{k}\Omega$: gain $=0.909$. |
| **Result** | $v_{out}/v_{cc}=r_O/(r_O+R_C)$, close to 1 |

**Problem 7**

| | BJT |
|---|---|
| **Problem** | A common-emitter amplifier has $R_C=10\,\mathrm{k}\Omega$, $V_{CC}=15\,\mathrm{V}$, $I_C=1\,\mathrm{mA}$, $\beta=100$ and $V_T=25\,\mathrm{mV}$. A small signal $v_{be}=5\sin\omega t\ \mathrm{mV}$ is applied. Find the gain, the collector voltage $v_C(t)$ and the base current $i_B(t)$. |
| **Given** | $V_{CC}=15\,\mathrm{V}$, $R_C=10\,\mathrm{k}\Omega$, $I_C=1\,\mathrm{mA}$, $\beta=100$, $v_{be}=5\sin\omega t\ \mathrm{mV}$ |
| **Answer 1** | **Transconductance.** $g_m=1/0.025=40\,\mathrm{mA/V}$. |
| **Answer 2** | **Gain.** $A_v$ $=-g_mR_C$ $=-40\cdot10$ $=-400\,\mathrm{V/V}$. |
| **Answer 3** | **DC collector voltage.** $V_C=15-1\cdot10=5\,\mathrm{V}$. |
| **Answer 4** | **Collector voltage.** $v_C$ $=5+A_vv_{be}$ $=5-400\cdot0.005\sin\omega t$ $=5-2\sin\omega t\ \mathrm{V}$. |
| **Answer 5** | **Base current.** $I_B=I_C/\beta=10\,\mu\mathrm{A}$; $i_b$ $=g_mv_{be}/\beta$ $=40\cdot0.005/100\,\mathrm{mA}$ $=2\sin\omega t\ \mu\mathrm{A}$. |
| **Answer 6** | **Total.** $i_B$ $=10+2\sin\omega t\ \mu\mathrm{A}$. |
| **Result** | $A_v=-400\,\mathrm{V/V}$, $v_C=5-2\sin\omega t\ \mathrm{V}$, $i_B$ $=10+2\sin\omega t\ \mu\mathrm{A}$ |

| **Problem** | An npn common-emitter amplifier with a collector resistor to $V_{CC}=5\,\mathrm{V}$ is biased at $V_{CE}=1\,\mathrm{V}$. Find its small-signal gain and the largest peak $v_{be}$ for which the transistor stays active ($V_{CE,sat}=0.3\,\mathrm{V}$, $V_T=25\,\mathrm{mV}$). |
| **Given** | $V_{CC}=5\,\mathrm{V}$, $V_{CE}=1\,\mathrm{V}$, $V_{CE,sat}=0.3\,\mathrm{V}$, $V_T=25\,\mathrm{mV}$ |
| **Answer 1** | **Gain.** $A_v$ $=-\dfrac{I_CR_C}{V_T}$ $=-\dfrac{V_{CC}-V_{CE}}{V_T}$ $=-\dfrac{4}{0.025}$ $=-160\,\mathrm{V/V}$ |

| **Answer 3** | **Input limit.** $\hat v_{be}$ $=\dfrac{0.7}{160}$ $=4.38\,\mathrm{mV}$ $\approx4.4\,\mathrm{mV}$ |
| **Result** | $A_v=-160\,\mathrm{V/V}$, $\hat v_{be,max}$ $\approx4.4\,\mathrm{mV}$ |

For each supply find (i) the largest possible gain magnitude, (ii) the gain when biased at $V_{CE}=V_{CC}/2$, and (iii) the gain when a $0.4\,\mathrm{V}$ negative output swing is required with maximum gain. |

| **Answer 4** | **(iii) Swing requirement.** Bias at $V_{CE}=0.3+0.4=0.7\,\mathrm{V}$: $\lvert A_v\rvert$ $=\dfrac{V_{CC}-0.7}{V_T}$ $=12,\;32,\;52,\;92\,\mathrm{V/V}$ |
| **Result** | $\lvert A_v\rvert$: max $28,48,68,108$; at $V_{CC}/2$: $20,30,40,60$; with $0.4\,\mathrm{V}$ swing: $12,32,52,92$ for $V_{CC}=1.0,1.5,2.0,3.0\,\mathrm{V}$ |

For bias currents $I_C$ $=0.5,\,1,\,2.5,\,4,\,4.5\,\mathrm{mA}$ find $V_{CE}$, the small-signal gain, and the maximum positive and negative output swing ($V_{CE,sat}=0.3\,\mathrm{V}$, $V_T=25\,\mathrm{mV}$). |

**Problem 12**

| | BJT |
|---|---|
| **Problem** | An npn transistor is used as a common-emitter amplifier with collector resistor $R_C$ from $V_{CC}$ to the collector, input $v_{BE}$ at the base and output $v_{CE}$ at the collector. The collector current includes the Early effect, $i_C$ $=I_Se^{v_{BE}/V_T}(1+v_{CE}/V_A)$. (a) Derive the small-signal gain $A_v=dv_{CE}/dv_{BE}$ at the bias point and show that $A_v$ $=-\dfrac{I_CR_C/V_T}{1+I_CR_C/(V_A+V_{CE})}$ $=-\dfrac{(V_{CC}-V_{CE})/V_T}{1+(V_{CC}-V_{CE})/(V_A+V_{CE})}$. (b) Evaluate it for $V_{CC}=5\,\mathrm{V}$, $V_{CE}=3\,\mathrm{V}$, $V_A=100\,\mathrm{V}$, with and without the Early effect. |
| **Given** | $V_{CC}=5\,\mathrm{V}$, $V_{CE}=3\,\mathrm{V}$, $V_A=100\,\mathrm{V}$, $V_T=25\,\mathrm{mV}$ |
| **Answer 1** | **Circuit equations.** $i_C$ $=I_Se^{v_{BE}/V_T}\left(1+\dfrac{v_{CE}}{V_A}\right)$ and $v_{CE}=V_{CC}-R_Ci_C$. |
| **Answer 2** | **Differentiate.** $A_v$ $=\dfrac{dv_{CE}}{dv_{BE}}$ $=-R_C\dfrac{di_C}{dv_{BE}}$ with $\dfrac{di_C}{dv_{BE}}$ $=\dfrac{I_Se^{V_{BE}/V_T}}{V_T}\left(1+\dfrac{V_{CE}}{V_A}\right)+\dfrac{I_Se^{V_{BE}/V_T}}{V_A}A_v$ $=\dfrac{I_C}{V_T}+\dfrac{I_Se^{V_{BE}/V_T}}{V_A}A_v$. |
| **Answer 3** | **Express $I_Se^{V_{BE}/V_T}$ with $I_C$.** At the bias point $I_C$ $=I_Se^{V_{BE}/V_T}(1+V_{CE}/V_A)$, so $I_Se^{V_{BE}/V_T}$ $=\dfrac{I_CV_A}{V_A+V_{CE}}$ and $\dfrac{I_Se^{V_{BE}/V_T}}{V_A}$ $=\dfrac{I_C}{V_A+V_{CE}}$. |
| **Answer 4** | **Solve for $A_v$.** $A_v$ $=-\dfrac{I_CR_C}{V_T}-\dfrac{I_CR_C}{V_A+V_{CE}}A_v\ \Rightarrow\ A_v\left(1+\dfrac{I_CR_C}{V_A+V_{CE}}\right)$ $=-\dfrac{I_CR_C}{V_T}$, i.e. $A_v$ $=-\dfrac{I_CR_C/V_T}{1+I_CR_C/(V_A+V_{CE})}$. |
| **Answer 5** | **Use the dc loop.** $I_CR_C=V_{CC}-V_{CE}$, which gives the second form of the result (Q.E.D.). |
| **Answer 6** | **Without the Early effect.** $I_CR_C=5-3=2\,\mathrm{V}$, so $A_v$ $=-\dfrac{2}{0.025}$ $=-80\,\mathrm{V/V}$. |
| **Answer 7** | **With the Early effect.** Denominator $1+\dfrac{2}{100+3}=1.0194$, so $A_v$ $=-\dfrac{80}{1.0194}$ $=-78.5\,\mathrm{V/V}$ (a 1.9 % reduction). |
| **Result** | $A_v$ $=-\dfrac{(V_{CC}-V_{CE})/V_T}{1+(V_{CC}-V_{CE})/(V_A+V_{CE})}$; $A_v=-80\,\mathrm{V/V}$ without and $-78.5\,\mathrm{V/V}$ with the Early effect |

Find $I_C$ and the gain, then compute the output change $v_o$ for $v_{BE}$ changed by $+5\,\mathrm{mV}$ and by $-5\,\mathrm{mV}$ (a) with the small-signal gain and (b) with the exponential characteristic, and compare. |
| **Given** | $V_{CC}=3\,\mathrm{V}$, $R_C=2\,\mathrm{k\Omega}$, $V_{CE}=1\,\mathrm{V}$, $V_T=25\,\mathrm{mV}$, $v_{be}=\pm5\,\mathrm{mV}$ |

| **Answer 5** | **Exponential, $-5\,\mathrm{mV}$.** $i_C$ $=I_Ce^{-5/25}$ $=0.819\,\mathrm{mA}$, so $\Delta i_C=-0.181\,\mathrm{mA}$ and $v_o=+363\,\mathrm{mV}$. |
| **Answer 6** | **Compare.** For $+5\,\mathrm{mV}$ the linear value ($-400\,\mathrm{mV}$) underestimates $\lvert v_o\rvert$ by $9.7\,\%$; for $-5\,\mathrm{mV}$ it overestimates by $10.3\,\%$, i.e. about $10\,\%$ either way. |
| **Result** | $I_C=1\,\mathrm{mA}$, $A_v=-80\,\mathrm{V/V}$; linear $\mp400\,\mathrm{mV}$ versus exponential $-443\,\mathrm{mV}$ / $+363\,\mathrm{mV}$ for $v_{be}=\pm5\,\mathrm{mV}$ |

**Problem 14**

| **Problem** | A common-source NMOS stage has input at the gate, a drain resistor $R_D$ to the supply, and its output at the drain. Find the short-circuit transconductance $G_m=i_{out}/v_{in}$ obtained by shorting the output node to ac ground. |
| **Given** | $g_{m1}$, $r_{O1}$, $R_D$ (symbolic); check values $g_m=2\,\mathrm{mS}$, $r_O=20\,\mathrm{k}\Omega$, $R_D=5\,\mathrm{k}\Omega$ |
| **Answer 1** | **Define the test.** $G_m$ $=\left.\dfrac{i_{out}}{v_{in}}\right\rvert_{v_{out}=0}$: force the drain to ac ground and measure the current flowing into the short. |
| **Answer 2** | **Current in $R_D$.** The top of $R_D$ is the supply (ac ground) and the bottom is the shorted output (ac ground), so $R_D$ carries no current. |
| **Answer 3** | **Current in $r_{O1}$.** The drain is at ac zero and the source is at ground, so $r_{O1}$ also has no voltage across it and carries no current. |
| **Answer 4** | **Short-circuit current.** All of the transistor current goes through the short: $i_{out}=g_{m1}v_{in}$ (direction as drawn by the published solution). |
| **Answer 5** | **Result.** $G_m=\dfrac{i_{out}}{v_{in}}=g_{m1}$: here the circuit transconductance equals that of the transistor. |
| **Answer 6** | **Cross-check.** With a small load $R_L$ the gain is $v_{out}/v_{in}$ $=-g_m(R_D\parallel r_O\parallel R_L)$, and $\lvert v_{out}/(R_Lv_{in})\rvert\to g_m$ as $R_L\to0$. |
| **Result** | $G_m=g_{m1}$ |

| **Answer 3** | **(b) Bias point.** $V_{GS}=1.5\,\mathrm{V}$, $I_D$ $=\tfrac12(1)(0.5)^2$ $=0.125\,\mathrm{mA}$, $V_{DS}=5-0.125(24)=2\,\mathrm{V}$ |
| **Answer 4** | **(b) Gain.** $A_v$ $=-k_nV_{OV}R_D$ $=-1(0.5)(24)$ $=-12\,\mathrm{V/V}$ |
| **Answer 5** | **(c) Input limit.** Saturation ends at point B, $v_{GS}=1.605\,\mathrm{V}$, so the largest input amplitude is $1.605-1.5=0.105\,\mathrm{V}$ ($\approx0.11\,\mathrm{V}$); $v_{GS}$ swings from $1.395$ to $1.605\,\mathrm{V}$ |
| **Answer 6** | **(c) Extremes.** $v_{GS}=1.395\,\mathrm{V}$: $i_D$ $=\tfrac12(0.395)^2$ $=0.0779\,\mathrm{mA}$, $v_{DS}$ $=5-0.0779(24)$ $=3.129\,\mathrm{V}$; $v_{GS}=1.605\,\mathrm{V}$: $i_D=0.1831\,\mathrm{mA}$, $v_{DS}=0.605\,\mathrm{V}$ |
| **Answer 7** | **(c) Large-signal gain.** $\dfrac{0.605-3.129}{1.605-1.395}$ $=-12.0\,\mathrm{V/V}$, equal to the small-signal gain: in saturation $v_{DS}$ $=V_{DD}-\tfrac12k_nR_D(v_{GS}-V_t)^2$ is a parabola, so a swing symmetric about $Q$ has a secant slope equal to the slope at $Q$. | **Result** | A$=(1\,\mathrm{V},5\,\mathrm{V})$; B$=(1.605\,\mathrm{V},0.605\,\mathrm{V})$; $I_D=0.125\,\mathrm{mA}$, $V_{DS}=2\,\mathrm{V}$, $A_v=-12\,\mathrm{V/V}$; input amplitude $0.105\,\mathrm{V}$; large-signal gain $-12.0\,\mathrm{V/V}$ |

| **Problem** | Four stages built on a pnp $Q_1$ whose emitter goes to $V_{CC}$ (through $R_E$, or through a device) and whose base takes $V_{in}$: (a) $R_E$ from $V_{CC}$ to the emitter; collector to the output node, then $R_C$ to a diode-connected npn $Q_2$ (base tied to collector, emitter grounded); output at the collector of $Q_1$. | **Given** | $V_A=\infty$; $R_E$, $R_C$; $g_{mj}$, $r_{\pi j}$ of each device |
| **Answer 1** | **Method.** A pnp has the same small-signal model as an npn, so each circuit is a degenerated common-emitter stage: $A_v$ $=-\dfrac{\beta R_L}{r_{\pi1}+(\beta+1)Z_E}$ $\approx-\dfrac{R_L}{1/g_{m1}+Z_E}$, $R_{in}=r_{\pi1}+(\beta+1)Z_E$, where $Z_E$ is the impedance from the emitter to ac ground and $R_L$ the impedance seen at the output node (the collector of $Q_1$ itself is an open circuit for $V_A=\infty$). |
| **Answer 2** | **Diode-connected device.** A transistor with base tied to collector looks like $\frac{1}{g_m}\parallel r_\pi$ $\approx\frac{1}{g_m}$. |

| **Answer 4** | **(b).** The signal current develops a voltage only across the diode: $A_v$ $=-\dfrac{1/g_{m2}}{1/g_{m1}+R_E}$, $R_{in}=r_{\pi1}+(\beta+1)R_E$, $R_{out}=\frac{1}{g_{m2}}$ ($R_C$ in series with the open collector of $Q_1$ does not count). |
| **Answer 5** | **(c).** $Z_E=\frac{1}{g_{m3}}$: $A_v$ $=-\dfrac{R_C+1/g_{m2}}{1/g_{m1}+1/g_{m3}}$, $R_{in}$ $=r_{\pi1}+\frac{\beta+1}{g_{m3}}$, $R_{out}=R_C+\frac{1}{g_{m2}}$. | **Answer 6** | **(d).** The base of $Q_2$ loads the output with $r_{\pi2}$: $R_L=R_C\parallel r_{\pi2}$, $A_v$ $=-\dfrac{R_C\parallel r_{\pi2}}{1/g_{m1}+R_E}$, $R_{in}=r_{\pi1}+(\beta+1)R_E$, $R_{out}=R_C\parallel r_{\pi2}$. |
| **Answer 7** | **Numerical check.** With all devices at $1\,\mathrm{mA}$, $\beta=100$, $R_E=200\,\Omega$, $R_C=1\,\mathrm{k}\Omega$ ($R_{in}=22.8\,\mathrm{k}\Omega$ in (a),(b),(d)): (a) $A_v=-4.50$, $R_{out}=1.03\,\mathrm{k}\Omega$; (b) $-0.113$, $25.7\,\Omega$; (c) $-19.7$ with $R_{in}=5.20\,\mathrm{k}\Omega$; (d) $-3.17$, $R_{out}=722\,\Omega$; the exact formulas equal a nodal solution. |
| **Result** | (a) $A_v$ $=-\frac{R_C+1/g_{m2}}{1/g_{m1}+R_E}$, $R_{out}=R_C+\frac{1}{g_{m2}}$; (b) $A_v=-\frac{1/g_{m2}}{1/g_{m1}+R_E}$, $R_{out}=\frac{1}{g_{m2}}$; (c) $A_v$ $=-\frac{R_C+1/g_{m2}}{1/g_{m1}+1/g_{m3}}$, $R_{in}$ $=r_{\pi1}+\frac{\beta+1}{g_{m3}}$, $R_{out}=R_C+\frac{1}{g_{m2}}$; (d) $A_v$ $=-\frac{R_C\parallel r_{\pi2}}{1/g_{m1}+R_E}$, $R_{out}=R_C\parallel r_{\pi2}$; $R_{in}=r_{\pi1}+(\beta+1)R_E$ in (a), (b), (d) |

| **Answer 2** | **Input resistance.** $R_{in}$ $=R_B\parallel(\beta+1)(r_e+R_e)$ $=100\parallel(101\times0.375)$ $=100\parallel37.9$ $=27.5\,\mathrm{k\Omega}$. |
| **Answer 3** | **Source to base.** $\dfrac{v_b}{v_{sig}}$ $=\dfrac{27.5}{27.5+10}$ $=0.733$. |
| **Answer 4** | **Base to output.** $\dfrac{v_o}{v_b}$ $=-\alpha\dfrac{R_C\parallel R_L}{r_e+R_e}$ $=-0.99\times\dfrac{5}{0.375}$ $=-13.2$. |
| **Answer 5** | **Overall gain.** $G_v$ $=-0.733\times13.2$ $=-9.68\,\mathrm{V/V}$ (about $-9.7$). |
| **Answer 6** | **Limit on the input.** $\hat v_b$ $=\hat v_{be}\dfrac{r_e+R_e}{r_e}$ $=5\times\dfrac{0.375}{0.125}$ $=15\,\mathrm{mV}$, so $\hat v_{sig}$ $=\dfrac{15}{0.733}$ $=20.5\,\mathrm{mV}$. |
| **Answer 7** | **Output amplitude.** $\hat v_o$ $=\lvert G_v\rvert\hat v_{sig}$ $=9.68\times20.5$ $=198\,\mathrm{mV}$ $\approx0.2\,\mathrm{V}$. |
| **Result** | $R_{in}=27.5\,\mathrm{k\Omega}$, $G_v=-9.7\,\mathrm{V/V}$, $\hat v_{sig}=20.5\,\mathrm{mV}$, $\hat v_o\approx0.2\,\mathrm{V}$ |

**Problem 27**

| | MOSFET |
|---|---|
| **Problem** | An NMOS transistor has its drain connected through $R_D$ to $V_{DD}$ and its source connected through $R_S$ to $V_{SS}$ (negative supply). The input $v_i$ is applied to the gate. For small signals (neglect $r_o$ and body effect) derive the gains $v_s/v_i$ at the source and $v_d/v_i$ at the drain. |
| **Given** | $R_D$, $R_S$ (symbolic), transconductance $g_m$, $r_o=\infty$ |
| **Answer 1** | **Input loop.** The gate-to-ground signal $v_i$ equals $v_{gs}$ plus the drop across $R_S$, and the source current is $g_mv_{gs}$: $v_i$ $=v_{gs}+g_mv_{gs}R_S$ $=g_mv_{gs}\left(\dfrac1{g_m}+R_S\right)$. |
| **Answer 2** | **Gate-source voltage.** $v_{gs}=\dfrac{v_i}{1+g_mR_S}$. |
| **Answer 3** | **Source voltage.** $v_s=g_mv_{gs}R_S$, hence $\dfrac{v_s}{v_i}$ $=\dfrac{g_mR_S}{1+g_mR_S}$ $=\dfrac{R_S}{1/g_m+R_S}$. |
| **Answer 4** | **Drain voltage.** The drain current $g_mv_{gs}$ flows down through $R_D$, so $v_d=-g_mv_{gs}R_D$ and $\dfrac{v_d}{v_i}$ $=-\dfrac{g_mR_D}{1+g_mR_S}$ $=-\dfrac{R_D}{1/g_m+R_S}$. |
| **Answer 5** | **Interpretation.** The source follows the gate (gain below 1) and the drain gain is the resistance ratio $R_D$ to $(1/g_m+R_S)$; for $g_mR_S\gg1$ it tends to $-R_D/R_S$. |
| **Result** | $\dfrac{v_s}{v_i}$ $=\dfrac{g_mR_S}{1+g_mR_S}$, $\dfrac{v_d}{v_i}$ $=-\dfrac{g_mR_D}{1+g_mR_S}$ |

| **Answer 3** | **$(W/L)_2$.** $100$ $=\tfrac12(128)\left(\dfrac WL\right)_2(0.15)^2\Rightarrow\left(\dfrac WL\right)_2$ $=69.4$. |
| **Answer 4** | **Small-signal parameters.** $g_{m1}$ $=2I_D/V_{OV}$ $=1.33\,\mathrm{mA/V}$; $r_{o1}$ $=5\times0.4/0.1$ $=20\,\mathrm{k}\Omega$; $r_{o2}$ $=6\times0.4/0.1$ $=24\,\mathrm{k}\Omega$. |
| **Answer 5** | **Gain.** $A_v$ $=-g_{m1}(r_{o1}\parallel r_{o2})$ $=-1.33\times(20\parallel24)$ $=-1.33\times10.9$ $=-14.5\,\mathrm{V/V}$. |

| **Problem** | An NMOS inverter has a driver $Q_1$ (input at its gate) and a load $Q_2$ whose gate is tied to $V_{DD}$ and whose source is the output node; both have the same $V_t$. (a) Derive $v_O$ as a function of $v_I$ while both devices are in saturation. (b) Find the gain for $(W/L)_1/(W/L)_2=10$. |
| **Given** | Identical $V_t$, $Q_2$ gate at $V_{DD}$; $(W/L)_1/(W/L)_2$ $=\dfrac{50/0.5}{5/0.5}$ $=10$ |
| **Answer 1** | **Load current.** $Q_2$ has $v_{GS2}=V_{DD}-v_O$: $i_D$ $=\tfrac12k_n^{\prime}(W/L)_2(V_{DD}-v_O-V_t)^2$ |

| **Answer 3** | **Equate the currents.** $(W/L)_2(V_{DD}-v_O-V_t)^2$ $=(W/L)_1(v_I-V_t)^2$ |

| **Result** | $v_O=V_{DD}-V_t+\sqrt{r}\,(V_t-v_I)$ with $r=(W/L)_1/(W/L)_2$; $A_v=-\sqrt r=-3.16$ for $r=10$ |

| **Answer 6** | **(d) KCL at X.** $-g_{m2}v_X-\frac{v_X}{r_{\pi2}}-g_{m1}v_{in}$ $=0$: $A_v$ $=-\dfrac{g_{m1}}{g_{m2}+1/r_{\pi2}}$, $R_{out}$ $=\frac{1}{g_{m2}}\parallel r_{\pi2}$ (again about $-1$ and $1/g_{m2}$). |

**Problem 38**

| | MOSFET |
|---|---|
| **Problem** | An NMOS common-source stage has an ideal current source as its load (infinite small-signal resistance), $M_1$ in saturation with finite output resistance $r_O=1/(\lambda I_D)$. Find the voltage gain and show how it changes with the channel length $L$, the bias current and the width staying fixed. |
| **Given** | $\lambda\ne0$, load resistance $=\infty$, $\lambda\propto1/L$ |
| **Answer 1** | **Gain with a finite load.** For a resistive load the gain is $A_v=-g_m(R_D\parallel r_O)$. |
| **Answer 2** | **Ideal current source.** $R_D\to\infty$ gives $A_v=-g_mr_O$, the largest gain one transistor can supply. |
| **Answer 3** | **Express in device terms.** $g_m$ $=\sqrt{2\mu_nC_{ox}\dfrac WL I_D}$ and $r_O=\dfrac{1}{\lambda I_D}$, so $\lvert A_v\rvert$ $=\dfrac{\sqrt{2\mu_nC_{ox}(W/L)I_D}}{\lambda I_D}$ $=\dfrac{1}{\lambda}\sqrt{\dfrac{2\mu_nC_{ox}W}{LI_D}}$. |
| **Answer 4** | **Include the $L$-dependence of $\lambda$.** Since $\lambda\propto1/L$, $\lvert A_v\rvert\propto L\cdot\dfrac{1}{\sqrt{L}}$ $=\sqrt L$ at fixed $W$ and $I_D$. |
| **Answer 5** | **Numerical illustration.** Take $W=10\,\mu\mathrm{m}$, $I_D=0.5\,\mathrm{mA}$, $\lambda=0.1\,\mathrm{V^{-1}}$ at $L=0.18\,\mu\mathrm{m}$: $\lvert A_v\rvert$ $=47.1,\ 66.7,\ 94.3$ for $L$ $=0.18,\ 0.36,\ 0.72\,\mu\mathrm{m}$, so doubling $L$ multiplies the gain by $\sqrt2$. |
| **Answer 6** | **Conclusion.** Although $g_m$ falls as $L$ grows, $r_O$ grows faster, so $\lvert A_v\rvert$ increases monotonically with $L$ (a longer device gives a higher intrinsic gain). |
| **Result** | $A_v=-g_mr_O$, $\lvert A_v\rvert\propto\sqrt{L}$ (increases with $L$) |

Find the dc $V_{GS}=V_{DS}$, $V_{OV}$, $g_m$, $r_o$ and the gain $v_o/v_i$. Then repeat if the current source is changed to $1\,\mathrm{mA}$ (same transistor). |

| **Answer 1** | **DC conditions.** No dc current flows in $R_G$, so $V_{GS}=V_{DS}=1\,\mathrm{V}$ and the transistor carries the full $I_D=0.2\,\mathrm{mA}$. $V_{OV}$ $=V_{GS}-V_t$ $=1-0.8$ $=0.2\,\mathrm{V}$, and $V_{DS}>V_{OV}$ so it is saturated. |
| **Answer 2** | **Small-signal parameters.** $g_m$ $=\dfrac{2I_D}{V_{OV}}$ $=\dfrac{2\times0.2}{0.2}$ $=2\,\mathrm{mA/V}$, $r_o$ $=\dfrac{V_A}{I_D}$ $=\dfrac{20}{0.2\,\mathrm{mA}}$ $=100\,\mathrm{k\Omega}$. |
| **Answer 3** | **Gain.** The input drives the gate; the drain sees $R_G$, $R_L$ and $r_o$ in parallel (the ideal current source is open): $\dfrac{v_o}{v_i}$ $=-g_m(R_G\parallel R_L\parallel r_o)$ $=-2\times(22\,\mathrm{M\Omega}\parallel15\,\mathrm{k\Omega}\parallel100\,\mathrm{k\Omega})$ $=-2\times13.04$ $=-26.1\,\mathrm{V/V}$. |
| **Answer 4** | **Current source of $1\,\mathrm{mA}$.** $I_D$ is 5 times larger, and $I_D\propto V_{OV}^2$ for a given device, so $V_{OV}=0.2\sqrt5=0.447\,\mathrm{V}$ and $V_{GS}$ $=V_{DS}$ $=0.8+0.447$ $=1.25\,\mathrm{V}$ (still saturated). |
| **Answer 5** | **New parameters and gain.** $g_m$ $=\dfrac{2\times1}{0.447}$ $=4.47\,\mathrm{mA/V}$, $r_o$ $=\dfrac{20}{1\,\mathrm{mA}}$ $=20\,\mathrm{k\Omega}$, $\dfrac{v_o}{v_i}$ $=-4.47\times(22\,\mathrm{M\Omega}\parallel15\,\mathrm{k\Omega}\parallel20\,\mathrm{k\Omega})$ $=-4.47\times8.57$ $=-38.3\,\mathrm{V/V}$. |
| **Result** | $200\,\mu\mathrm{A}$: $V_{OV}=0.2\,\mathrm{V}$, $g_m=2\,\mathrm{mA/V}$, $r_o=100\,\mathrm{k\Omega}$, $v_o/v_i=-26.1\,\mathrm{V/V}$; $1\,\mathrm{mA}$: $V_{OV}=0.447\,\mathrm{V}$, $g_m=4.47\,\mathrm{mA/V}$, $r_o=20\,\mathrm{k\Omega}$, $v_o/v_i=-38.3\,\mathrm{V/V}$ |

**Problem 47**

| | MOSFET |
|---|---|
| **Problem** | In the circuit an NMOS transistor $Q_1$ receives the input $v_i$ at its gate; its drain is loaded by a diode-connected PMOS transistor $Q_2$. A PMOS transistor $Q_3$ shares the gate and source of $Q_2$, so it mirrors the drain current of $Q_1$, and $Q_3$ delivers its current into a load resistor $R_L$ to ground. Neglecting $r_o$ of all devices, derive the voltage gain $v_o/v_i$. |
| **Given** | Symbolic: $g_{m1}$, $R_L$, $(W/L)_2$, $(W/L)_3$ (same $L$) |
| **Answer 1** | **Load of $Q_1$.** Diode-connected $Q_2$ presents $1/g_{m2}$ to small signals. |
| **Answer 2** | **Voltage at the drain of $Q_1$.** $v_{d1}$ $=-g_{m1}v_i\cdot\dfrac1{g_{m2}}$ $=-\dfrac{g_{m1}}{g_{m2}}v_i$, which is also the gate voltage of $Q_3$ referred to its source. |
| **Answer 3** | **Mirror action.** $Q_3$ copies the signal current of $Q_2$ in the ratio of the aspect ratios: $i_{d3}$ $=g_{m1}v_i\dfrac{(W/L)_3}{(W/L)_2}$ $=g_{m1}v_i\dfrac{W_3}{W_2}$ (equal $L$). |
| **Answer 4** | **Output voltage.** $v_o$ $=i_{d3}R_L$ $=g_{m1}v_iR_L\dfrac{W_3}{W_2}$. |
| **Answer 5** | **Gain.** $\dfrac{v_o}{v_i}$ $=g_{m1}R_L\dfrac{W_3}{W_2}$. |
| **Result** | $\dfrac{v_o}{v_i}$ $=g_{m1}R_L\,\dfrac{W_3}{W_2}$ |

| | MOSFET |

Find (a) the dc input voltage and the $W/L$ ratios, (b) the voltage gain, (c) the allowed output range, and (d) the supply voltage needed if $Q_2$, $Q_3$ are replaced by a resistor $R_D$ that keeps the same gain and output swing. |

| **Answer 7** | **(d) Resistor load.** Same gain needs $R_D=r_{o2}=24\,\mathrm{k}\Omega$. Same swing means biasing $Q_1$ at the middle of $0.15$ to $1.15\,\mathrm{V}$: $V_{DS1}=0.65\,\mathrm{V}$. |
| **Answer 8** | **(d) Supply.** $V_{DD}$ $=V_{DS1}+I_DR_D$ $=0.65+0.1\times24$ $=3.05\,\mathrm{V}$, much more than the $0.13\,\mu\mathrm{m}$ technology allows. |
| **Result** | $V_I=0.55\,\mathrm{V}$, $(W/L)_1=17.4$, $(W/L)_{2,3}=69.4$; $A_v=-14.5$; $0.15\,\mathrm{V}\le v_O\le1.15\,\mathrm{V}$; $V_{DD}=3.05\,\mathrm{V}$ |

| **Given** | $g_m=2\,\mathrm{mA/V}$, $r_o=20\,\mathrm{k}\Omega$, $R_s=20\,\mathrm{k}\Omega$, $R_L=20$, $40$, $200\,\mathrm{k}\Omega$ |
| **Answer 1** | **(a) Current gain.** The whole source current of the transistor flows through $R_L$, so $i_o$ is the share of $i_{sig}$ entering the source: $i_o=i_{sig}\dfrac{R_s}{R_s+R_{in}}$, i.e. $\dfrac{i_o}{i_{sig}}$ $=\dfrac{R_s}{R_s+R_{in}}$. |
| **Answer 2** | **(b) $R_L=20\,\mathrm{k}\Omega$.** $R_{in}$ $=\dfrac{r_o+R_L}{g_mr_o+1}$ $=\dfrac{40}{41}$ $=0.976\,\mathrm{k}\Omega$; $\dfrac{i_o}{i_{sig}}$ $=\dfrac{20}{20.976}$ $=0.953$. |
| **Answer 3** | **(c) $R_L=40\,\mathrm{k}\Omega$.** $R_{in}$ $=\dfrac{60}{41}$ $=1.46\,\mathrm{k}\Omega$; $\dfrac{i_o}{i_{sig}}$ $=\dfrac{20}{21.46}$ $=0.932$. A $100\%$ rise of $R_L$ changes the gain by only about $2\%$. |
| **Answer 4** | **(d) $R_L=200\,\mathrm{k}\Omega$.** $R_{in}$ $=\dfrac{220}{41}$ $=5.37\,\mathrm{k}\Omega$; $\dfrac{i_o}{i_{sig}}$ $=\dfrac{20}{25.37}$ $=0.788$: a $900\%$ rise of $R_L$ costs about $17\%$ of the gain. |
| **Answer 5** | **(e) Comment.** The current gain changes little for large changes of $R_L$: the CG stage acts as a current buffer. |
| **Result** | $R_{in}=0.976$, $1.46$, $5.37\,\mathrm{k}\Omega$; $i_o/i_{sig}=0.953$, $0.932$, $0.788$ |

| **Result** | (a) $A_v=g_mR_C=38.5$, $R_{in}\approx1/g_m=13.0\,\Omega$, $R_{out}=500\,\Omega$; (b) lowest $V_B$ the input allows and $R_C=(V_{CC}-V_B)/I_C$, giving $A_v=(V_{CC}-V_B)/V_T$ |

**Problem 59**

| | BJT |
|---|---|
| **Problem** | Find the voltage gain of four common-base stages with $V_A=\infty$, the input applied at the emitter: (a) npn $Q_1$ with base on $V_b$, collector (output) connected through $R_C$ to the emitter of a diode-connected npn $Q_2$ whose base and collector are on $V_{CC}$. (b) pnp $Q_1$ with emitter driven by $V_{in}$, base on $V_b$, collector (output) driving the base of an npn $Q_2$ whose collector is on $V_{CC}$ and emitter grounded. (c) npn $Q_1$ with base on $V_b$, $R_E$ from the emitter (input) to ground, collector (output) connected through $R_C$ to a diode-connected pnp $Q_3$ whose emitter is on $V_{CC}$. (d) The circuit of (c) with the input applied through a series resistor $R_S$ to the emitter. |
| **Given** | $V_A=\infty$; $R_C$, $R_E$, $R_S$; $g_{mj}$, $r_{\pi j}$ |
| **Answer 1** | **Common-base core.** With the base at ac ground and an ideal source at the emitter, $v_{BE}=-v_{in}$, so $i_C=-g_{m1}v_{in}$ and $v_{out}=g_{m1}v_{in}R_L$: $A_v=g_{m1}R_L$, where $R_L$ is the impedance at the collector node. |
| **Answer 2** | **(a).** $R_L$ $=R_C+\frac{1}{g_{m2}}\parallel r_{\pi2}$ $\approx R_C+\frac{1}{g_{m2}}$: $A_v$ $=g_{m1}\left(R_C+\frac{1}{g_{m2}}\right)$ $\approx g_{m1}R_C+1$, since $Q_2$ carries the collector current of $Q_1$. |
| **Answer 3** | **(b) Load.** The only load is the base of $Q_2$: $R_L=r_{\pi2}$, so $A_v=g_{m1}r_{\pi2}$. |
| **Answer 4** | **(b) Bias relation.** The collector current of $Q_1$ is the base current of $Q_2$: $I_{C2}=\beta I_{C1}$, so $r_{\pi2}$ $=\dfrac{\beta V_T}{\beta I_{C1}}$ $=\dfrac{1}{g_{m1}}$ and $A_v=1$. |
| **Answer 5** | **(c).** $R_E$ is driven directly by the ideal source and does not affect the gain: $A_v$ $=g_{m1}\left(R_C+\frac{1}{g_{m3}}\parallel r_{\pi3}\right)$ $\approx g_{m1}\left(R_C+\frac{1}{g_{m3}}\right)$. |
| **Answer 6** | **(d) Input divider.** The emitter node sees $R_E\parallel\frac{1}{g_{m1}}$ to ground and $R_S$ to the source: $\dfrac{v_E}{v_{in}}$ $=\dfrac{R_E\parallel(1/g_{m1})}{R_S+R_E\parallel(1/g_{m1})}$. |
| **Answer 7** | **(d) Gain.** $A_v$ $=\dfrac{R_E\parallel(1/g_{m1})}{R_S+R_E\parallel(1/g_{m1})}\,g_{m1}\left(R_C+\frac{1}{g_{m3}}\right)$ $=\dfrac{g_{m1}R_E\left(R_C+1/g_{m3}\right)}{R_S+R_E+g_{m1}R_SR_E}$. |
| **Answer 8** | **Numerical check.** With $1\,\mathrm{mA}$ in $Q_1$ (and $Q_2$, $Q_3$ in (a), (c), (d)), $\beta=100$, $R_C=1\,\mathrm{k}\Omega$, $R_E=500\,\Omega$, $R_S=200\,\Omega$: (a) $39.5$, (b) $1.00$, (c) $39.5$, (d) $4.30$, each equal to a nodal solution (with $r_\pi/(\beta+1)$ in place of $1/g_m$). |
| **Result** | (a) $A_v$ $=g_{m1}(R_C+1/g_{m2})$ $\approx g_{m1}R_C+1$; (b) $A_v=g_{m1}r_{\pi2}=1$; (c) $A_v=g_{m1}(R_C+1/g_{m3})$; (d) $A_v$ $=\dfrac{g_{m1}R_E(R_C+1/g_{m3})}{R_S+R_E+g_{m1}R_SR_E}$ |

**Problem 60**

| | BJT |

| **Problem** | Find the input impedance $R_{in}$ looking into the emitter of npn $Q_1$ (collector to $V_{CC}$ through $R_1$) in four circuits with $V_A=\infty$: (a) the base of $Q_1$ is driven by the emitter of a diode-connected npn $Q_2$ whose base and collector are on $V_{CC}$; (b) the base of $Q_1$ is on a node pulled up to $V_{CC}$ by $R_2$ and pulled down by the collector of an npn $Q_2$ (base on $V_b$, emitter grounded); (c) as (b) but $Q_2$ is diode-connected (base tied to its collector); (d) the base node, pulled up by $R_2$, also drives the base of an npn $Q_2$ whose emitter is grounded and collector is on $V_{CC}$. |
| **Given** | $V_A=\infty$; $R_1$, $R_2$; $g_{mj}$, $r_{\pi j}$, $\beta$ |
| **Answer 1** | **Looking into an emitter.** With an impedance $Z_B$ from the base to ac ground: $R_{in}$ $=\dfrac{Z_B+r_{\pi1}}{\beta+1}$ $\approx\dfrac{1}{g_{m1}}+\dfrac{Z_B}{\beta+1}$; $R_1$ does not matter for $V_A=\infty$. |
| **Answer 2** | **(a) Base network.** $Z_B$ $=\frac{1}{g_{m2}}\parallel r_{\pi2}$. Here $Q_2$ carries only the base current of $Q_1$: $I_{E2}=I_{B1}$, so $\frac{1}{g_{m2}}$ $=\frac{\beta+1}{g_{m1}}$ and $Z_B$ $=\frac{1}{g_{m2}}\cdot\frac{\beta}{\beta+1}$ $=r_{\pi1}$. |
| **Answer 3** | **(a) Result.** $R_{in}$ $=\dfrac{2r_{\pi1}}{\beta+1}$ $\approx\dfrac{2}{g_{m1}}$. |
| **Answer 4** | **(b).** The collector of $Q_2$ is an open circuit ($V_A=\infty$): $Z_B=R_2$, $R_{in}$ $=\dfrac{R_2+r_{\pi1}}{\beta+1}$ $\approx\dfrac{1}{g_{m1}}+\dfrac{R_2}{\beta+1}$. |
| **Answer 5** | **(c).** Diode-connected $Q_2$: $Z_B$ $=R_2\parallel\frac{1}{g_{m2}}\parallel r_{\pi2}$, $R_{in}$ $=\dfrac{R_2\parallel\frac{1}{g_{m2}}\parallel r_{\pi2}+r_{\pi1}}{\beta+1}$. |
| **Answer 6** | **(d).** The base of $Q_2$ adds $r_{\pi2}$: $Z_B=R_2\parallel r_{\pi2}$, $R_{in}$ $=\dfrac{R_2\parallel r_{\pi2}+r_{\pi1}}{\beta+1}$. |
| **Answer 7** | **Numerical check.** With $I_{C1}=1\,\mathrm{mA}$, $\beta=100$, $R_1=R_2=2\,\mathrm{k}\Omega$ and $1\,\mathrm{mA}$ in $Q_2$ for (c), (d): (a) $51.5\,\Omega$, (b) $45.5\,\Omega$, (c) $26.0\,\Omega$, (d) $36.9\,\Omega$, each equal to a nodal solution. |
| **Result** | (a) $R_{in}$ $=\frac{2r_{\pi1}}{\beta+1}$ $\approx\frac{2}{g_{m1}}$; (b) $\frac{R_2+r_{\pi1}}{\beta+1}$; (c) $\frac{R_2\parallel(1/g_{m2})\parallel r_{\pi2}+r_{\pi1}}{\beta+1}$; (d) $\frac{R_2\parallel r_{\pi2}+r_{\pi1}}{\beta+1}$ |

| | BJT |
|---|---|
| **Problem** | A common-base stage: npn $Q_1$ with its base on $V_b$, the input applied directly at the emitter, and an ideal current source from $V_{CC}$ as the collector load (output at the collector). With a finite Early voltage ($r_O$), find the voltage gain and the input and output impedances. |
| **Given** | Ideal current-source load; $g_m$, $r_\pi$, $r_O$ |
| **Answer 1** | **KCL at the output.** The ideal source carries no signal current, so the small-signal collector current is zero: $g_mv_{BE}+\dfrac{v_{out}-v_{in}}{r_O}$ $=0$ with $v_{BE}=-v_{in}$. |

| **Answer 4** | **Output impedance.** Set $v_{in}=0$: the emitter is grounded, $v_{BE}=0$, and only $r_O$ remains: $R_{out}=r_O$. |
| **Answer 5** | **Numerical check.** With $I_C=1\,\mathrm{mA}$, $\beta=100$, $V_A=10\,\mathrm{V}$: $A_v=386$, $R_{in}=2.60\,\mathrm{k}\Omega$, $R_{out}=10.0\,\mathrm{k}\Omega$, the same as a nodal solution. |
| **Result** | $A_v=1+g_mr_O$, $R_{in}=r_\pi$ (infinite if base current is neglected), $R_{out}=r_O$ |

| | BJT |

| **Problem** | A common-base stage: npn $Q_1$ with its base on $V_b$, an ideal current source from $V_{CC}$ as the collector load (output at the collector), and the input $V_{in}$ applied to the emitter through a source resistance $R_S$. **Problem 64**

| | BJT |
|---|---|
| **Problem** | A common-base stage has an npn transistor with collector resistor $R_C$ to $V_{CC}$ (output at the collector), a resistance $R_B$ from the base to ac ground, and the input $v_{in}$ applied to the emitter through a degeneration resistor $R_E$. With a finite Early voltage ($r_O$) and $\beta\gg1$, use the small-signal model to find the voltage gain. |
| **Given** | $R_C$, $R_E$, $R_B$, $g_m$, $r_\pi$, $r_O$; $\beta\gg1$ |
| **Answer 1** | **Base loop.** The base current flows through $R_B$: $v_B=-i_BR_B$ and $v_\pi=v_B-v_E=i_Br_\pi$, so $v_\pi=-\dfrac{r_\pi}{r_\pi+R_B}v_E$. Define $g$ $=\dfrac{g_mr_\pi}{r_\pi+R_B}$ $=\dfrac{\beta}{r_\pi+R_B}$, so $g_mv_\pi=-g\,v_E$. |
| **Answer 2** | **KCL at the collector.** $-\dfrac{v_{out}}{R_C}$ $=g_mv_\pi+\dfrac{v_{out}-v_E}{r_O}$ gives $v_{out}$ $=v_E\,\dfrac{(1+gr_O)R_C}{R_C+r_O}$. |
| **Answer 3** | **KCL at the emitter.** The current from $R_E$ leaves through the base path and the collector: $\dfrac{v_{in}-v_E}{R_E}$ $=\dfrac{v_E}{r_\pi+R_B}+\dfrac{v_{out}}{R_C}$. |
| **Answer 4** | **Eliminate $v_E$.** $\dfrac{v_{in}}{R_E}$ $=v_E\left[\dfrac{1}{R_E}+\dfrac{1}{r_\pi+R_B}+\dfrac{1+gr_O}{R_C+r_O}\right]$. |
| **Answer 5** | **Exact gain.** $A_v$ $=\dfrac{(1+gr_O)R_C}{(R_C+r_O)\left(1+\dfrac{R_E}{r_\pi+R_B}\right)+(1+gr_O)R_E}$. |
| **Answer 6** | **With $\beta\gg1$.** The base current through $R_E$ is negligible ($R_E\ll r_\pi+R_B$) and $\dfrac{1}{g}$ $\approx\dfrac{1}{g_m}+\dfrac{R_B}{\beta}$: $A_v$ $\approx\dfrac{(1+gr_O)R_C}{R_C+r_O+(1+gr_O)R_E}$. |
| **Answer 7** | **Limit $r_O\to\infty$.** $A_v\to\dfrac{R_C}{R_E+\frac{1}{g_m}+\frac{R_B}{\beta}}$, the standard result for a common-base stage with base resistance. |
| **Answer 8** | **Numerical check.** With $I_C=1\,\mathrm{mA}$, $\beta=100$, $V_A=10\,\mathrm{V}$, $R_C=2\,\mathrm{k}\Omega$, $R_E=200\,\Omega$, $R_B=5\,\mathrm{k}\Omega$: exact $A_v=6.83$, nodal $6.83$; the $\beta\gg1$ form gives $6.88$ and the $r_O\to\infty$ limit $7.25$ ($r_O=10\,\mathrm{k}\Omega$ is only five times $R_C$ here). |
| **Result** | $A_v$ $=\dfrac{(1+gr_O)R_C}{(R_C+r_O)\left(1+\frac{R_E}{r_\pi+R_B}\right)+(1+gr_O)R_E}$ $\approx\dfrac{(1+gr_O)R_C}{R_C+r_O+(1+gr_O)R_E}$, $g$ $=\dfrac{\beta}{r_\pi+R_B}$ $\approx\left(\dfrac{1}{g_m}+\dfrac{R_B}{\beta}\right)^{-1}$ |

| **Answer 4** | **Output resistance: structure.** Set $v_{in}=0$. Looking into the drain of $M_1$, the source node has the resistance $\dfrac{1}{g_{m2}}\parallel r_{O2}\parallel R_S$ to ground. |
| **Answer 5** | **Degenerated-stage formula.** $R_{out1}$ $=(1+g_{m1}r_{O1})\left(\dfrac{1}{g_{m2}}\parallel r_{O2}\parallel R_S\right)+r_{O1}$ $\approx g_{m1}r_{O1}\left(\dfrac{1}{g_{m2}}\parallel R_S\right)+r_{O1}$. |
| **Answer 6** | **Include $R_D$.** $R_{out}=R_{out1}\parallel R_D$. |
| **Answer 7** | **Numerical check.** $g_{m1}=3\,\mathrm{mS}$, $g_{m2}=2\,\mathrm{mS}$, $R_S=500\,\Omega$, $R_D=4\,\mathrm{k}\Omega$: nodal analysis and the formula give $A_v=3.43$. | **Result** | $A_v$ $=\dfrac{g_{m1}R_D}{1+(g_{m1}+g_{m2})R_S}$; $R_{out}$ $=\left[(1+g_{m1}r_{O1})\left(\dfrac{1}{g_{m2}}\parallel r_{O2}\parallel R_S\right)+r_{O1}\right]\parallel R_D$ |

| **Answer 2** | **(a) Gain.** $\dfrac{v_{out}}{v_{cc}}$ $=\dfrac{r_O}{r_O+R_C}$ |
| **Answer 3** | **(b) Resistance below the drain.** With the gate at ac ground, the degenerated transistor looks like $R_{down}=r_O+(1+g_mr_O)R_S$ from the drain. |
| **Answer 4** | **(b) Gain.** $\dfrac{v_{out}}{v_{dd}}$ $=\dfrac{r_O+(1+g_mr_O)R_S}{r_O+(1+g_mr_O)R_S+R_D}$ |
| **Answer 5** | **(c) KCL at the emitter.** $v_{BE}=-v_{out}$; the ideal $I_{EE}$ carries no small-signal current, so $-g_mv_{out}-\dfrac{v_{out}}{r_\pi}+\dfrac{v_{cc}-v_{out}}{r_O}$ $=0$. |
| **Answer 6** | **(c) Gain.** $\dfrac{v_{out}}{v_{cc}}$ $=\dfrac{1/r_O}{g_m+1/r_\pi+1/r_O}$ $\approx\dfrac{1}{g_mr_O}$ |
| **Answer 7** | **(d) KCL at the source.** $v_{GS}=-v_{out}$: $-g_mv_{out}+\dfrac{v_{dd}-v_{out}}{r_O}$ $=\dfrac{v_{out}}{R_S}$. |
| **Answer 8** | **(d) Gain.** $\dfrac{v_{out}}{v_{dd}}$ $=\dfrac{R_S}{R_S+r_O+g_mr_OR_S}$ $\approx\dfrac{1}{g_mr_O}$ for $g_mR_S\gg1$ |
| **Answer 9** | **Numeric check.** With $g_m=40\,\mathrm{mS}$, $r_\pi=2.5\,\mathrm{k}\Omega$, $r_O=10\,\mathrm{k}\Omega$, $R_C=R_D=1\,\mathrm{k}\Omega$, $R_S=500\,\Omega$: (a) $0.909$, (b) $0.995$, (c) $2.47\times10^{-3}$, (d) $2.38\times10^{-3}$; nodal analysis gives the same. |
| **Result** | (a) $r_O/(r_O+R_C)$; (b) $R_{down}/(R_{down}+R_D)$ with $R_{down}=r_O+(1+g_mr_O)R_S$; (c) $(1/r_O)/(g_m+1/r_\pi+1/r_O)$ $\approx1/(g_mr_O)$; (d) $R_S/(R_S+r_O+g_mr_OR_S)$ |

(a) NMOS follower $M_1$ (drain at $V_{DD}$, input at the gate, output at the source); from the output, $R_S$ in series with the drain of NMOS current source $M_2$ (gate at $V_b$, source grounded). (c) Same follower; load is a PMOS $M_2$ with its source at the output and both gate and drain grounded. (d) Same follower; load is NMOS $M_2$ (drain at the output, source grounded) whose gate is driven by the divider $R_1$ (output to gate) and $R_2$ (gate to ground). (e) PMOS follower $M_2$ (input at the gate, drain grounded, source at the output), loaded by a PMOS current source $M_3$ (gate at $V_b$) from $V_{DD}$ and by an NMOS $M_1$ with drain and gate at $V_{DD}$ and source at the output. (f) PMOS follower $M_1$ (input at the gate, drain grounded, source at the output) fed from $V_{DD}$ by a cascode PMOS current source $M_3$ (gate $V_{b2}$) over $M_2$ (gate $V_{b1}$, drain at the output). |
| **Given** | $\lambda\ne0$: $g_{mi}$, $r_{Oi}$; $R_S$, $R_1$, $R_2$ symbolic |
| **Answer 1** | **Method.** A follower with total resistance $Z$ from its source to ac ground has $A_v=\dfrac{Z}{Z+1/g_m}$, with $Z$ including the device own $r_O$ (its drain sits at ac ground) |
| **Answer 2** | **(a).** $Z=r_{O1}\parallel(R_S+r_{O2})$, $A_v=\dfrac{Z}{Z+1/g_{m1}}$ |
| **Answer 3** | **(b).** Looking into the drain of degenerated $M_2$: $r_{O2}+(1+g_{m2}r_{O2})R_S$; $Z$ $=r_{O1}\parallel[r_{O2}+(1+g_{m2}r_{O2})R_S]$, $A_v=\dfrac{Z}{Z+1/g_{m1}}$ |
| **Answer 4** | **(c).** $M_2$ has $v_{SG2}=v_{out}$ and a grounded drain, so it presents $\dfrac{1}{g_{m2}}\parallel r_{O2}$: $Z$ $=r_{O1}\parallel r_{O2}\parallel\dfrac{1}{g_{m2}}$, $A_v=\dfrac{Z}{Z+1/g_{m1}}$ |
| **Answer 5** | **(d).** $v_{G2}=\dfrac{R_2}{R_1+R_2}v_{out}$, so $M_2$ draws $g_{m2}\dfrac{R_2}{R_1+R_2}v_{out}$: $Z$ $=r_{O1}\parallel r_{O2}\parallel(R_1+R_2)\parallel\dfrac{R_1+R_2}{g_{m2}R_2}$, $A_v=\dfrac{Z}{Z+1/g_{m1}}$ |
| **Answer 6** | **(e).** Follower device is $M_2$; $M_1$ is diode-connected ($1/g_{m1}\parallel r_{O1}$), $M_3$ gives $r_{O3}$: $Z$ $=\dfrac{1}{g_{m1}}\parallel r_{O1}\parallel r_{O2}\parallel r_{O3}$, $A_v$ $=\dfrac{Z}{Z+1/g_{m2}}$ $=\dfrac{g_{m2}}{g_{m1}+g_{m2}+r_{O1}^{-1}+r_{O2}^{-1}+r_{O3}^{-1}}$ |
| **Answer 7** | **(f).** Cascode load seen from the output: $r_{O2}+(1+g_{m2}r_{O2})r_{O3}$; $Z$ $=r_{O1}\parallel[r_{O2}+(1+g_{m2}r_{O2})r_{O3}]$, $A_v=\dfrac{Z}{Z+1/g_{m1}}$ |
| **Answer 8** | **Check.** Nodal analysis with $g_{m1}=4\,\mathrm{mS}$, $g_{m2}=3\,\mathrm{mS}$, $r_{O1}=20\,\mathrm{k}\Omega$, $r_{O2}=15\,\mathrm{k}\Omega$, $r_{O3}=25\,\mathrm{k}\Omega$, $R_S=1\,\mathrm{k}\Omega$, $R_1=10\,\mathrm{k}\Omega$, $R_2=5\,\mathrm{k}\Omega$ matches all six formulas |
| **Result** | $A_v=\dfrac{Z}{Z+1/g_m}$ with (a) $Z=r_{O1}\parallel(R_S+r_{O2})$; (b) $r_{O1}\parallel[r_{O2}+(1+g_{m2}r_{O2})R_S]$; (c) $r_{O1}\parallel r_{O2}\parallel g_{m2}^{-1}$; (d) $r_{O1}\parallel r_{O2}\parallel(R_1+R_2)\parallel\dfrac{R_1+R_2}{g_{m2}R_2}$; (e) $g_{m1}^{-1}\parallel r_{O1}\parallel r_{O2}\parallel r_{O3}$ with $g_m=g_{m2}$; (f) $r_{O1}\parallel[r_{O2}+(1+g_{m2}r_{O2})r_{O3}]$ |

| | MOSFET |
|---|---|
| **Problem** | Two NMOS transistors are stacked: $M_1$ on top (output at its drain) and $M_2$ below it with its source grounded; each gate is held at its own fixed bias, and both are saturated. Find the small-signal output resistance at the drain of $M_1$ and compare with the diode-connected case of Problem 80. |
| **Given** | $g_{m1}$, $r_{O1}$, $r_{O2}$ symbolic, $g_{m1}r_{O2}\gg1$ |
| **Answer 1** | **Role of $M_2$.** Its gate-source voltage is fixed, so it acts as a current source, i.e. the resistance $r_{O2}$ from the source of $M_1$ to ground. |
| **Answer 2** | **Degenerated-stage formula.** $R_{out}=(1+g_{m1}r_{O1})R_S+r_{O1}$ with $R_S=r_{O2}$. |
| **Answer 3** | **Result.** $R_{out}$ $=(1+g_{m1}r_{O1})r_{O2}+r_{O1}$ $\approx g_{m1}r_{O1}r_{O2}+r_{O1}$. |

| **Answer 5** | **Check the assumption.** $r_\pi$ $=\beta/g_m$ $=2.6\,\mathrm{k}\Omega\gg R_E$ $=26\,\Omega$, so $R_E\parallel r_\pi\approx R_E$ is justified. |
| **Answer 6** | **Exact solution of the full formula.** With $x=R_E\parallel r_\pi$, $(1+g_mx)r_O+x=2r_O$ gives $x$ $=\dfrac{r_O}{1+g_mr_O}$ $=25.9\,\Omega$, so $R_E$ $=\dfrac{x\,r_\pi}{r_\pi-x}$ $=26.2\,\Omega$. | **Result** | $R_E\approx 26\,\Omega$ ($=1/g_m$; exact formula gives $26.2\,\Omega$) |

**Problem 99**

| | BJT |
|---|---|
| **Problem** | Folk wisdom says the output impedance of a common-base (CB) stage is much higher than that of a common-emitter (CE) stage. It is "demonstrated" by two tests. Test (a): the emitter is grounded, a constant current $I_B$ is forced into the base, and the collector voltage $V_1$ is swept; $I_C$ rises with slope $1/r_O$. Test (b): the base is at a fixed voltage, a constant current $I_E$ is pulled out of the emitter, and $V_1$ is swept; $I_C$ is almost flat. Explain why these tests do not reflect practical circuits. (Numerical illustration, not part of the original question: $I_C=1\,\mathrm{mA}$, $V_A=10\,\mathrm{V}$, $\beta=100$.) |
| **Given** | $I_C=1\,\mathrm{mA}$, $V_A=10\,\mathrm{V}$, $\beta=100$, $V_T=26\,\mathrm{mV}$, $V_A$ finite |
| **Answer 1** | **Small-signal model of each test.** An ideal constant current source has infinite small-signal resistance, so it becomes an open circuit: in (a) the base is open and the emitter grounded; in (b) the base is at ac ground and the emitter is open. |
| **Answer 2** | **Test (a).** The emitter is grounded and no current flows in $r_\pi$, so $v_\pi=0$ and $g_mv_\pi=0$. Only $r_O$ remains: $R_{out}$ $=r_O$ $=\dfrac{V_A}{I_C}$ $=10\,\mathrm{k}\Omega$. |
| **Answer 3** | **Test (b).** This is an emitter-degenerated stage with $R_E=\infty$: $R_{out}$ $=[1+g_m(R_E\parallel r_\pi)]r_O+(R_E\parallel r_\pi)\to(1+g_mr_\pi)r_O+r_\pi$ $=(\beta+1)r_O+r_\pi$ $\approx\beta r_O$. |
| **Answer 4** | **Numbers.** $g_m=38.5\,\mathrm{mS}$, $r_\pi=2.6\,\mathrm{k}\Omega$, so $R_{out}$ $=101\times10\,\mathrm{k}\Omega+2.6\,\mathrm{k}\Omega$ $\approx1.01\,\mathrm{M}\Omega$, about $100$ times larger than in (a). |
| **Answer 5** | **Why the tests mislead.** The large value in (b) exists only because the emitter is driven by an ideal current source (infinite resistance). In practice stages are driven by voltage sources with finite resistance $R_S$. With the input set to zero for the output-resistance calculation, the CB stage becomes a transistor with base grounded and emitter resistance $R_S$, which is the same circuit as a CE stage with emitter degeneration $R_S$. The two stages then have the same $R_{out}$ $=[1+g_m(R_S\parallel r_\pi)]r_O+(R_S\parallel r_\pi)$ (equal to $r_O$ when $R_S=0$), so the claim of higher CB output impedance is irrelevant. |
| **Result** | (a) $R_{out}=r_O=10\,\mathrm{k}\Omega$; (b) $R_{out}$ $\approx(\beta+1)r_O+r_\pi$ $\approx1.01\,\mathrm{M}\Omega$; both differ only because the input sources are ideal current sources, so the comparison does not apply to practical (voltage-driven) stages |

| **Answer 2** | **(b) Emitter impedance.** Into the emitter of $Q_2$ (base and collector at ac ground): $Z_E$ $=\dfrac{1}{g_{m2}}\parallel r_{\pi2}\parallel r_{O2}$. |
| **Answer 3** | **(b) Degenerated collector.** $R_{out}$ $=\left[1+g_{m1}(Z_E\parallel r_{\pi1})\right]r_{O1}+Z_E\parallel r_{\pi1}$. |
| **Answer 4** | **(b) Equal currents.** $I_{E2}=I_{E1}$, so $Z_E\approx1/g_{m1}$ and $R_{out}\approx2r_{O1}$. |
| **Answer 5** | **Numerical check.** With $I_C=1\,\mathrm{mA}$, $V_A=5\,\mathrm{V}$: $9.90\,\mathrm{k}\Omega$ by formula and nodal analysis ($2r_{O1}=10\,\mathrm{k}\Omega$). |
| **Result** | (a) $R_C\parallel r_{O1}$; (b) $[1+g_{m1}(Z_E\parallel r_{\pi1})]r_{O1}+Z_E\parallel r_{\pi1}$, $Z_E$ $=\frac{1}{g_{m2}}\parallel r_{\pi2}\parallel r_{O2}$, $\approx2r_{O1}$ |

**Problem 105**

| | BJT |
|---|---|
| **Problem** | Find the impedance $R_{out}$ looking into the collector of an npn $Q_1$ whose base is at ac ground, for three emitter networks: (a) the emitter of $Q_1$ goes to the emitter of a pnp $Q_2$ whose base and collector are grounded; (b) the emitter of $Q_1$ is pulled down by an ideal current source $I_1$ and also drives the emitter of an npn $Q_2$ whose collector is on $V_{CC}$ and whose base goes to ac ground through $R_B$; (c) the emitter of $Q_1$ goes to ground through $R_1$ and drives the base of an npn $Q_2$ whose emitter is grounded and collector is on $V_{CC}$. Assume $\beta\gg1$ and a finite Early voltage. |
| **Given** | $r_{O1}$, $r_{O2}$, $g_{m1}$, $g_{m2}$, $r_{\pi1}$, $r_{\pi2}$, $R_B$, $R_1$; $\beta\gg1$ |
| **Answer 1** | **General result.** For a transistor with grounded base and an impedance $R_X$ from emitter to ac ground (Problem 92): $R_{out}$ $=\left[1+g_{m1}(R_X\parallel r_{\pi1})\right]r_{O1}+R_X\parallel r_{\pi1}$. Only $R_X$ differs between the three circuits. |
| **Answer 2** | **(a) Emitter network.** $Q_2$ (pnp, base and collector grounded) is diode-connected: $R_X$ $=\frac{1}{g_{m2}}\parallel r_{\pi2}\parallel r_{O2}$ $\approx\frac{1}{g_{m2}}$. |
| **Answer 3** | **(a) Result.** $R_{out}$ $\approx\left(1+\dfrac{g_{m1}}{g_{m2}}\right)r_{O1}+\dfrac{1}{g_{m2}}$. Both devices carry the same current, so $g_{m1}=g_{m2}$ and $R_{out}\approx2r_{O1}$. |
| **Answer 4** | **(b) Emitter network.** $I_1$ is ideal (open). Looking into the emitter of $Q_2$: $\dfrac{R_B+r_{\pi2}}{\beta+1}\parallel r_{O2}$ $\approx\dfrac{1}{g_{m2}}+\dfrac{R_B}{\beta}$, so $R_X$ $\approx\dfrac{1}{g_{m2}}+\dfrac{R_B}{\beta}$. |
| **Answer 5** | **(b) Result.** $R_X\ll r_{\pi1}$ here (for $\beta\gg1$ and $R_B\ll\beta r_{\pi1}$), so $r_{\pi1}$ drops out of the parallel combination: $R_{out}$ $\approx\left[1+g_{m1}\left(\dfrac{1}{g_{m2}}+\dfrac{R_B}{\beta}\right)\right]r_{O1}+\dfrac{1}{g_{m2}}+\dfrac{R_B}{\beta}$. |
| **Answer 6** | **(c) Emitter network.** The base of $Q_2$ adds $r_{\pi2}$ in parallel with $R_1$: $R_X=R_1\parallel r_{\pi2}$. |
| **Answer 7** | **(c) Result.** $R_{out}$ $=\left[1+g_{m1}(R_1\parallel r_{\pi1}\parallel r_{\pi2})\right]r_{O1}+R_1\parallel r_{\pi1}\parallel r_{\pi2}$ $\approx\left[1+g_{m1}(R_1\parallel r_{\pi1}\parallel r_{\pi2})\right]r_{O1}$. Here $r_{\pi1}$ has to stay: $R_1\parallel r_{\pi2}$ is in general not small compared with $r_{\pi1}$. Only if $R_1\ll r_{\pi1},r_{\pi2}$ does the result reduce to $(1+g_{m1}R_1)r_{O1}$. |
| **Answer 8** | **Numerical check.** With $1\,\mathrm{mA}$ in each device, $\beta=100$, $V_A=10\,\mathrm{V}$, $R_B=5\,\mathrm{k}\Omega$, $R_1=2\,\mathrm{k}\Omega$: (a) $19.8\,\mathrm{k}\Omega$ ($\approx2r_{O1}$ $=20.0\,\mathrm{k}\Omega$); (b) $38.0\,\mathrm{k}\Omega$ (approximate form $39.3\,\mathrm{k}\Omega$); (c) $314\,\mathrm{k}\Omega$ (compact form $313\,\mathrm{k}\Omega$; leaving $r_{\pi1}$ out would give $445\,\mathrm{k}\Omega$, 42 % too high). The exact expressions equal a nodal solution. |
| **Result** | (a) $R_{out}$ $\approx(1+g_{m1}/g_{m2})r_{O1}$ $\approx2r_{O1}$; (b) $R_{out}$ $\approx\left[1+g_{m1}\left(\frac{1}{g_{m2}}+\frac{R_B}{\beta}\right)\right]r_{O1}$; (c) $R_{out}$ $\approx\left[1+g_{m1}(R_1\parallel r_{\pi1}\parallel r_{\pi2})\right]r_{O1}$ (each plus the small emitter impedance $R_X\parallel r_{\pi1}$) |

| **Given** | $V_{CC}=2.5\,\mathrm{V}$, $R_B=100\,\mathrm{k}\Omega$, $R_C=1\,\mathrm{k}\Omega$, speaker dc resistance $\le1\,\Omega$ (taken as $1\,\Omega$) |

| **Answer 2** | **Highest possible collector voltage.** Even with $Q_1$ completely off, $R_C$ and the speaker form a divider: $V_C\le\dfrac{1\,\Omega}{1\,\mathrm{k}\Omega+1\,\Omega}\times2.5\,\mathrm{V}$ $=2.5\,\mathrm{mV}$. |
| **Answer 3** | **Region of operation.** The base is at $V_{BE}\approx0.7$ to $0.8\,\mathrm{V}$ while the collector is at no more than a few millivolts, so $V_{BC}\approx V_{BE}-V_C\approx0.7$ to $0.8\,\mathrm{V}>0$: the collector-base junction is forward biased and $Q_1$ is in deep saturation, not in the active region. |
| **Answer 4** | **Conclusion.** The collector cannot take any signal swing (it is held at about $0\,\mathrm{V}$), and the transistor has no gain. The speaker must be coupled through a capacitor, as in the next example. |
| **Result** | The speaker shorts the collector to ground at dc, so $V_C\approx0$ and $Q_1$ is in deep saturation; the circuit has no gain |

**Problem 113**

| | BJT |
|---|---|
| **Problem** | A common-emitter amplifier has $R_B=100\,\mathrm{k}\Omega$ from $V_{CC}=2.5\,\mathrm{V}$ to the base, $R_C=1\,\mathrm{k}\Omega$ from $V_{CC}$ to the collector, an input coupling capacitor, and an output coupling capacitor $C_2$ to an $8\,\Omega$ speaker. The measured dc collector voltage is $1.5\,\mathrm{V}$ (active region). With $I_S=5\times10^{-17}\,\mathrm{A}$ and $V_A=\infty$: (a) find $\beta$; (b) explain why there is no gain by finding $\lvert A_v\rvert$. |
| **Given** | $V_{CC}=2.5\,\mathrm{V}$, $R_B=100\,\mathrm{k}\Omega$, $R_C=1\,\mathrm{k}\Omega$, $R_{sp}=8\,\Omega$, $V_C=1.5\,\mathrm{V}$, $I_S=5\times10^{-17}\,\mathrm{A}$, $V_T=26\,\mathrm{mV}$ |
| **Answer 1** | **(a) Collector current.** The drop across $R_C$ is $2.5-1.5=1.0\,\mathrm{V}$, so $I_C$ $=\dfrac{1\,\mathrm{V}}{1\,\mathrm{k}\Omega}$ $=1\,\mathrm{mA}$ (the capacitors carry no dc). |
| **Answer 2** | **(a) Base-emitter voltage.** $V_{BE}$ $=V_T\ln\dfrac{I_C}{I_S}$ $=26\,\mathrm{mV}\times\ln(2\times10^{13})$ $=796\,\mathrm{mV}$. |
| **Answer 3** | **(a) Base current and $\beta$.** $I_B$ $=\dfrac{V_{CC}-V_{BE}}{R_B}$ $=\dfrac{2.5-0.796}{100\,\mathrm{k}\Omega}$ $=17.0\,\mu\mathrm{A}$, so $\beta=\dfrac{I_C}{I_B}=58.7$. |
| **Answer 4** | **(b) Small-signal parameter.** $g_m$ $=\dfrac{I_C}{V_T}$ $=38.5\,\mathrm{mS}$. |
| **Answer 5** | **(b) Load seen by the collector.** For signals, $C_2$ is a short, so the $8\,\Omega$ speaker is in parallel with $R_C$: $R_C\parallel R_{sp}$ $=1000\parallel8$ $=7.94\,\Omega$. |
| **Answer 6** | **(b) Gain.** $\lvert A_v\rvert$ $=g_m(R_C\parallel R_{sp})$ $=38.5\,\mathrm{mS}\times7.94\,\Omega$ $=0.305$: the stage attenuates. Its output resistance ($\approx R_C=1\,\mathrm{k}\Omega$) is far larger than the load, so almost all of the signal is lost across $R_C$. |
| **Result** | (a) $\beta\approx58.7$; (b) $\lvert A_v\rvert\approx0.305<1$ because $R_C\parallel R_{sp}$ $\approx8\,\Omega$ |

**Problem 114**

| | BJT |
|---|---|
| **Problem** | A common-base stage has its base held at a dc voltage $V_b$ and a resistor $R_C$ from the collector to $V_{CC}$. To keep the signal source from disturbing the bias, the source $v_{in}$ is connected to the emitter through a series coupling capacitor $C_1$, and nothing else is connected to the emitter. Explain why the circuit does not work. |
| **Given** | $V_b$ fixed, $R_C$, coupling capacitor $C_1$ at the emitter, no other emitter connection |
| **Answer 1** | **DC analysis.** For bias, $C_1$ is an open circuit. The emitter is then connected to nothing, so there is no dc path for the emitter current: $I_E=0$. |
| **Answer 2** | **Collector current.** $I_C=\alpha I_E=0$ (apart from leakage), hence $V_{out}=V_{CC}-R_CI_C=V_{CC}$. |
| **Answer 3** | **Small-signal consequence.** $g_m=I_C/V_T=0$, so $A_v=g_mR_C=0$: the stage has no gain. |
| **Answer 4** | **Analogy.** This is the common-base counterpart of the common-emitter circuit with no base-current path: the terminal that supplies the current needs a dc connection (here a resistor to ground or a current source). |
| **Result** | No dc path for the emitter current, so $I_E=I_C=0$, $g_m=0$ and $A_v=0$ |

**Problem 115**

| | BJT |
|---|---|
| **Problem** | To fix the missing bias path of the previous common-base circuit (base at a dc voltage $V_b$, collector resistor $R_C$ to $V_{CC}$, signal $v_{in}$ coupled through $C_1$), a student ties the emitter directly to ground so that $V_{BE}=V_b$ and a collector current can flow. Explain why this also fails. |
| **Given** | $V_b$ fixed, $R_C$ to $V_{CC}$, emitter wired to ground at the node where the capacitor $C_1$ ends |
| **Answer 1** | **DC effect.** With the emitter at ground, $V_{BE}=V_b$ and $I_C=I_S\exp(V_b/V_T)$ flows, so the bias point can be set (a suitable $V_b$ is needed to avoid saturation). |
| **Answer 2** | **Small-signal effect.** The emitter node is a fixed $0\,\mathrm{V}$ for all signals as well: it is shorted to ground, so the signal source (behind $C_1$) is also shorted and cannot change the emitter voltage. |

| **Answer 4** | **Moral.** The emitter must be neither open (no bias) nor shorted to ground (no signal); it needs a bias element such as a resistor $R_E$ to ground or a current source. |

**Problem 116**

| | BJT |
|---|---|
| **Problem** | A common-emitter stage: the source $v_{in}$ with series resistance $R_S$ drives the base directly. A resistor $R_1$ connects the base to a node $M$, and $R_2$ connects $M$ to the collector; $M$ is tied to ground by a very large capacitor $C_1$. | **Answer 3** | **Gain.** Common base driven by an ideal source: $A_v$ $=g_{m1}R_L$ $=g_{m1}\left(R_1\parallel\frac{1}{g_{m2}}\right)$. Since $Q_2$ carries $I_{C1}$ plus the current of $R_1$, $g_{m2}>g_{m1}$ and $A_v<1$. |
| **Answer 4** | **Input impedance.** $R_{in}$ $=\dfrac{r_{\pi1}}{\beta+1}$ $\approx\dfrac{1}{g_{m1}}$. |
| **Answer 5** | **Output impedance.** The collector of $Q_1$ is open ($V_A=\infty$): $R_{out}$ $=R_1\parallel\frac{1}{g_{m2}}\parallel r_{\pi2}$. |
| **Answer 6** | **Numerical check.** With $I_{C1}=1\,\mathrm{mA}$, $I_{C2}=1.2\,\mathrm{mA}$, $\beta=100$, $R_1=R_2=5\,\mathrm{k}\Omega$: $A_v=0.822$, $R_{in}=25.7\,\Omega$, $R_{out}=21.4\,\Omega$, equal to a nodal solution. |
| **Result** | $A_v$ $=g_{m1}\left(R_1\parallel\frac{1}{g_{m2}}\right)$, $R_{in}\approx\frac{1}{g_{m1}}$, $R_{out}$ $=R_1\parallel\frac{1}{g_{m2}}$ |

| **Given** | $V_G=2\,\mathrm{V}$, $V_D=2.5\,\mathrm{V}$, $I_D=0.25\,\mathrm{mA}$, $V_t=0.75\,\mathrm{V}$, $V_A=50\,\mathrm{V}$, $R_{sig}=100\,\mathrm{k\Omega}$, $R_D=R_L=10\,\mathrm{k\Omega}$ |
| **Answer 1** | **(a) Bias.** $V_{GS}=2-1=1\,\mathrm{V}$, $V_{OV}=1-0.75=0.25\,\mathrm{V}$. $V_D$ $=2.5\,\mathrm{V}>V_G-V_t$ $=1.25\,\mathrm{V}$, so the device is in saturation. |
| **Answer 2** | **(a) $k_n$.** $I_D=\tfrac12k_nV_{OV}^2$: $0.25=\tfrac12k_n(0.25)^2$, so $k_n=8\,\mathrm{mA/V^2}$. |
| **Answer 3** | **(b) Input resistance.** $R_{in}$ $=R_{G1}\parallel R_{G2}$ $=750\parallel500$ $=300\,\mathrm{k\Omega}$. |
| **Answer 4** | **(b) Small-signal parameters.** $g_m$ $=\dfrac{2I_D}{V_{OV}}$ $=2\,\mathrm{mA/V}$; $r_o$ $=\dfrac{V_A}{I_D}$ $=200\,\mathrm{k\Omega}$. |
| **Answer 5** | **(b) Gain.** $\lvert A_v\rvert$ $=g_m(r_o\parallel R_D\parallel R_L)$ $=2\times4.88$ $=9.76$; $G_v$ $=-\dfrac{R_{in}}{R_{in}+R_{sig}}\lvert A_v\rvert$ $=-0.75\times9.76$ $=-7.3\,\mathrm{V/V}$. |
| **Answer 6** | **(c) Saturation limit.** $v_D$ $=2.5-9.76\hat v_{gs}\ge v_G-V_t$ $=2+\hat v_{gs}-0.75$ gives $\hat v_{gs}$ $=\dfrac{1.25}{10.76}$ $=0.116\,\mathrm{V}$. |
| **Answer 7** | **(c) Signal and output.** $\hat v_{sig}$ $=\hat v_{gs}\dfrac{300+100}{300}$ $=0.155\,\mathrm{V}$; output amplitude $7.32\times0.155=1.13\,\mathrm{V}$. |
| **Answer 8** | **(d) Unbypassed $R_s$.** To double $\hat v_{sig}$ with the same $\hat v_{gs}$ we need $1+g_mR_s=2$, so $R_s=\dfrac{1}{g_m}=500\,\Omega$; $\hat v_{gs}$ and hence $\hat v_o\approx1.13\,\mathrm{V}$ are unchanged. |
| **Result** | $k_n=8\,\mathrm{mA/V^2}$; $g_m=2\,\mathrm{mA/V}$, $r_o=200\,\mathrm{k\Omega}$, $G_v=-7.3\,\mathrm{V/V}$; $\hat v_{sig}=0.155\,\mathrm{V}$ (output $1.13\,\mathrm{V}$); $R_s=500\,\Omega$ |

**Problem 86**

| | MOSFET |
|---|---|
| **Problem** | An NMOS transistor has a $10\,\mathrm{M\Omega}$ resistor $R_G$ from the gate to ground (dc), a source resistor $R_S$ to $-5\,\mathrm{V}$ and a drain resistor $R_D$ to $+5\,\mathrm{V}$. $k_n=5\,\mathrm{mA/V^2}$, $V_t=0.8\,\mathrm{V}$, $V_A=40\,\mathrm{V}$, $I_D=0.4\,\mathrm{mA}$. (e) With X grounded, a current source $i_{sig}=50\,\mu\mathrm{A}$ with $R_{sig}=100\,\mathrm{k\Omega}$ drives Z; find $v_y$ (no load, neglect $r_o$). |
| **Given** | $\pm5\,\mathrm{V}$, $I_D=0.4\,\mathrm{mA}$, $k_n=5\,\mathrm{mA/V^2}$, $V_t=0.8\,\mathrm{V}$, $V_A=40\,\mathrm{V}$, $R_G=10\,\mathrm{M\Omega}$ |
| **Answer 1** | **(a) Overdrive.** $I_D=\tfrac12k_nV_{OV}^2$: $0.4=2.5V_{OV}^2$, $V_{OV}=0.4\,\mathrm{V}$, $V_{GS}=0.8+0.4=1.2\,\mathrm{V}$. |
| **Answer 2** | **(a) $R_S$.** $V_G=0$, $V_S=-1.2\,\mathrm{V}$: $R_S$ $=\dfrac{-1.2-(-5)}{0.4}$ $=9.5\,\mathrm{k\Omega}$. |
| **Answer 3** | **(a) $R_D$.** Saturation needs $V_D\ge V_G-V_t=-0.8\,\mathrm{V}$; to allow a 0.8 V negative swing set $V_D=0$: $R_D$ $=\dfrac{5-0}{0.4}$ $=12.5\,\mathrm{k\Omega}$. |
| **Answer 4** | **(b) $g_m$, $r_o$.** $g_m$ $=\dfrac{2I_D}{V_{OV}}$ $=2\,\mathrm{mA/V}$; $r_o$ $=\dfrac{40}{0.4}$ $=100\,\mathrm{k\Omega}$. |
| **Answer 5** | **(c) CS gain.** $\dfrac{v_y}{v_{sig}}$ $=-\dfrac{R_G}{R_G+R_{sig}}g_m(r_o\parallel R_D\parallel R_L)$ $=-\dfrac{10}{11}\times2\times5.26$ $=-9.6\,\mathrm{V/V}$. |

**Problem 88**

| | MOSFET |
|---|---|
| **Problem** | An NMOS common-source amplifier is biased from $V_{DD}=10\,\mathrm{V}$ by feedback: a $2\,\mathrm{M\Omega}$ resistor $R_2$ from the drain to the gate and a $0.5\,\mathrm{M\Omega}$ resistor $R_1$ from the gate to ground (the signal source drives the gate through $R_1$, whose lower end is the signal source at dc ground); the drain resistor is $R_D$. $V_t=0.6\,\mathrm{V}$, $k_n=5\,\mathrm{mA/V^2}$, $V_A=60\,\mathrm{V}$, $V_{OV}=0.2\,\mathrm{V}$. (a) Find $R_D$. (b) Find $g_m$ and $r_o$. (c) Derive $\dfrac{v_o}{v_{sig}}$ $=-\dfrac{R_2/R_1}{1+\dfrac{1+R_2/R_1}{g_m(R_D\parallel r_o\parallel R_2)(1-1/(g_mR_2))}}$ and evaluate it. |
| **Given** | $V_{DD}=10\,\mathrm{V}$, $R_1=0.5\,\mathrm{M\Omega}$, $R_2=2\,\mathrm{M\Omega}$, $V_t=0.6\,\mathrm{V}$, $k_n=5\,\mathrm{mA/V^2}$, $V_A=60\,\mathrm{V}$, $V_{OV}=0.2\,\mathrm{V}$ |
| **Answer 1** | **Gate voltage.** $V_{GS}=V_t+V_{OV}=0.8\,\mathrm{V}$. With no gate current, $V_{GS}$ $=V_D\dfrac{R_1}{R_1+R_2}$ $=0.2V_D$, so $V_D=4\,\mathrm{V}$. |
| **Answer 2** | **Drain current.** $I_D$ $=\tfrac12k_nV_{OV}^2\left(1+\dfrac{V_{DS}}{V_A}\right)$ $=\tfrac12\times5\times0.04\times\left(1+\dfrac{4}{60}\right)$ $=0.1067\,\mathrm{mA}$. |
| **Answer 3** | **(a) $R_D$.** The divider draws $\dfrac{4}{2.5\,\mathrm{M\Omega}}$ $=1.6\,\mu\mathrm{A}$, so $R_D$ $=\dfrac{10-4}{0.1067+0.0016}$ $=55.4\,\mathrm{k\Omega}$. |
| **Answer 4** | **(b) $g_m$, $r_o$.** $g_m$ $=\dfrac{2I_D}{V_{OV}}$ $=1.07\,\mathrm{mA/V}$; $r_o$ $=\dfrac{V_A}{I_D}$ $=562\,\mathrm{k\Omega}$. |
| **Answer 5** | **(c) Node equation at the drain.** $\dfrac{v_o}{R_D}+\dfrac{v_o}{r_o}+\dfrac{v_o-v_{gs}}{R_2}+g_mv_{gs}$ $=0\Rightarrow v_o$ $=-A\,v_{gs}$ with $A$ $=g_m(R_D\parallel r_o\parallel R_2)\left(1-\dfrac{1}{g_mR_2}\right)$. |
| **Answer 6** | **(c) Gate voltage by superposition.** $v_{gs}$ $=v_{sig}\dfrac{R_2}{R_1+R_2}+v_o\dfrac{R_1}{R_1+R_2}$. Substituting: $v_o\left(1+A\dfrac{R_1}{R_1+R_2}\right)$ $=-A\dfrac{R_2}{R_1+R_2}v_{sig}$, i.e. $\dfrac{v_o}{v_{sig}}$ $=-\dfrac{R_2/R_1}{1+(1+R_2/R_1)/A}$ (Q.E.D.). |
| **Answer 7** | **(c) Numbers.** $A$ $=1.067\times(55.4\parallel562\parallel2000)\times(1-0.0005)$ $=52.5$; $\dfrac{R_2}{R_1}=4$: $\dfrac{v_o}{v_{sig}}$ $=-\dfrac{4}{1+5/52.5}$ $=-3.65\,\mathrm{V/V}$, close to the op-amp value $-R_2/R_1=-4$. |

| **Problem** | An NMOS common-source amplifier is biased by an ideal $200\,\mu\mathrm{A}$ current source into the drain (supplying the transistor current). | **Given** | $I=200\,\mu\mathrm{A}$, $R_{G2}=3\,\mathrm{M}\Omega$ (drain-gate), $R_{G1}=2\,\mathrm{M}\Omega$ (gate-ground), $k_n'(W/L)=2\,\mathrm{mA/V^2}$, $V_t=0.5\,\mathrm{V}$, $V_A=20\,\mathrm{V}$ |

| **Answer 2** | **Largest gain with the swing.** Gain is largest for the smallest $V_{DS}$. At the negative peak $v_{DS}=V_{DS}-0.2$ must stay at or above $V_{OV}$, so $V_{DS}=V_{OV}+0.2$. (1) |
| **Answer 3** | **Gain in terms of $V_{OV}$.** $\lvert A_v\rvert$ $=g_mR_D$ $=\dfrac{2I_D}{V_{OV}}R_D$ $=\dfrac{V_{DD}-V_{DS}}{0.5V_{OV}}$, so $10$ $=\dfrac{1.8-V_{DS}}{0.5V_{OV}}\Rightarrow V_{DS}$ $=1.8-5V_{OV}$. (2) |

| **Answer 5** | **Drain current.** $I_D$ $=\dfrac{V_{DD}-V_{DS}}{R_D}$ $=\dfrac{1.8-0.467}{10}$ $=0.133\,\mathrm{mA}$. |
| **Answer 6** | **Device size.** $k_n$ $=\dfrac{2I_D}{V_{OV}^2}$ $=\dfrac{2\times0.133}{0.267^2}$ $=3.75\,\mathrm{mA/V^2}$ (equivalently $g_m/V_{OV}$), so $\dfrac WL$ $=\dfrac{k_n}{k_n^\prime}$ $=\dfrac{3.75}{0.38}$ $=9.87$. |
| **Result** | $V_{OV}=0.267\,\mathrm{V}$, $V_{DS}=0.467\,\mathrm{V}$, $I_D=0.133\,\mathrm{mA}$, $W/L\approx9.9$ |

| **Result** | (a) $A_v$ $=-\beta(R_1+\frac{1}{g_{m2}}\parallel r_{\pi2})/[r_{\pi1}+(\beta+1)R_E]$, $R_{in}=r_{\pi1}+(\beta+1)R_E$, $R_{out}$ $=R_1+\frac{1}{g_{m2}}\parallel r_{\pi2}$; (b),(c) $A_v$ $=-\beta R_C/[r_{\pi1}+(\beta+1)(\frac{1}{g_{m2}}\parallel r_{\pi2})]$, $R_{in}$ $=r_{\pi1}+(\beta+1)(\frac{1}{g_{m2}}\parallel r_{\pi2})$, $R_{out}=R_C$; (d) as (c) with $R_B$ added to the denominator and to $R_{in}$; (e) $A_v$ $=-\beta R_C/(R_B+r_{\pi1}+r_{\pi2})$, $R_{in}=R_B+r_{\pi1}+r_{\pi2}$, $R_{out}=R_C$ |

| | BJT |

| **Problem** | A Darlington pair: npn $Q_1$ takes the input at its base and its emitter drives the base of npn $Q_2$; both collectors are tied to $V_{CC}$, and $V_A=\infty$. The emitter current of $Q_1$ is the base current of $Q_2$, $I_{E1}=I_{B2}=I_{C2}/\beta$. (a) With the emitter of $Q_2$ grounded, find the impedance seen at the base of $Q_1$. (b) With the base of $Q_1$ grounded, find the impedance seen at the emitter of $Q_2$. (c) Find the current gain $(I_{C1}+I_{C2})/I_{B1}$. |
| **Given** | $V_A=\infty$, collectors on $V_{CC}$, $I_{E1}=I_{B2}$, equal $\beta$ |
| **Answer 1** | **Bias relation.** $I_{C1}$ $=\frac{\beta}{\beta+1}I_{E1}$ $=\dfrac{I_{C2}}{\beta+1}$, so $r_{\pi1}$ $=\dfrac{\beta V_T}{I_{C1}}$ $=(\beta+1)r_{\pi2}$ and $g_{m1}=\dfrac{g_{m2}}{\beta+1}$. |
| **Answer 2** | **(a) Load of $Q_1$.** The emitter of $Q_1$ sees the base of $Q_2$ (emitter grounded): $Z_E=r_{\pi2}$. |
| **Answer 3** | **(a) Result.** $R_{in}$ $=r_{\pi1}+(\beta+1)r_{\pi2}$ $=2(\beta+1)r_{\pi2}$ $\approx2\beta r_{\pi2}$ $=\dfrac{2\beta^2}{g_{m2}}$. |
| **Answer 4** | **(b) Impedance at the emitter of $Q_1$.** With its base grounded: $\dfrac{r_{\pi1}}{\beta+1}=r_{\pi2}$, which is now the base resistance of $Q_2$. |
| **Answer 5** | **(b) Result.** $R_{out}$ $=\dfrac{r_{\pi2}+r_{\pi2}}{\beta+1}$ $=\dfrac{2r_{\pi2}}{\beta+1}$ $\approx\dfrac{2}{g_{m2}}$. |
| **Answer 6** | **(c) Currents.** $I_{C1}=\beta I_{B1}$, $I_{B2}=I_{E1}=(\beta+1)I_{B1}$, $I_{C2}=\beta(\beta+1)I_{B1}$. |
| **Answer 7** | **(c) Current gain.** $\dfrac{I_{C1}+I_{C2}}{I_{B1}}$ $=\beta+\beta(\beta+1)$ $=\beta(\beta+2)$ $\approx\beta^2$. |
| **Answer 8** | **Numerical check.** With $I_{C2}=1\,\mathrm{mA}$, $\beta=100$: $r_{\pi2}=2.60\,\mathrm{k}\Omega$, $R_{in}=525\,\mathrm{k}\Omega$, $R_{out}=51.5\,\Omega$, current gain $10200$; nodal solutions agree. |
| **Result** | (a) $R_{in}$ $=2(\beta+1)r_{\pi2}$ $\approx2\beta r_{\pi2}$; (b) $R_{out}$ $=\frac{2r_{\pi2}}{\beta+1}$ $\approx\frac{2}{g_{m2}}$; (c) $\beta(\beta+2)\approx\beta^2$ |

| **Answer 3** | **Output at $Y$ (degenerated CS).** $M_1$ is a CS stage with source resistance $\dfrac{1}{g_{m2}}\parallel r_{O2}$ and drain load $\dfrac{1}{g_{m3}}\parallel r_{O3}\parallel r_{O4}$ ($M_3$ diode, $M_4$ current source). |
| **Answer 4** | **Gain to $Y$.** $\dfrac{v_Y}{v_{in}}$ $=-\dfrac{\frac{1}{g_{m3}}\parallel r_{O3}\parallel r_{O4}}{\frac{1}{g_{m1}}+\frac{1}{g_{m2}}\parallel r_{O2}}$. |
| **Answer 5** | **Numerical check.** $g_{m1}=4$, $g_{m2}=2$, $g_{m3}=1\,\mathrm{mS}$, $r_{O2}=30$, $r_{O3}=25$, $r_{O4}=35\,\mathrm{k}\Omega$: nodal analysis and the formulas give $v_X/v_{in}=0.663$, $v_Y/v_{in}=-1.26$. |
| **Result** | $\dfrac{v_X}{v_{in}}$ $=\dfrac{\frac{1}{g_{m2}}\parallel r_{O2}}{\frac{1}{g_{m2}}\parallel r_{O2}+\frac{1}{g_{m1}}}$; $\dfrac{v_Y}{v_{in}}$ $=-\dfrac{\frac{1}{g_{m3}}\parallel r_{O3}\parallel r_{O4}}{\frac{1}{g_{m1}}+\frac{1}{g_{m2}}\parallel r_{O2}}$ |

**Problem 132**

| | MOSFET |
|---|---|
| **Problem** | A common-source NMOS $M_1$ (input at the gate, source grounded) has a drain resistor $R_{D1}$ to $V_{DD}$; its drain, node $X$, drives the source of a common-gate NMOS $M_2$ whose gate is at $V_b$ and whose drain has $R_{D2}$ to $V_{DD}$ and is the output. With $\lambda=0$: (a) find $v_{out}/v_{in}$ $=(v_X/v_{in})(v_{out}/v_X)$; (b) simplify for $R_{D1}\to\infty$ and explain the result. |
| **Given** | $\lambda=0$; $g_{m1}$, $g_{m2}$, $R_{D1}$, $R_{D2}$ symbolic |
| **Answer 1** | **Impedance at X.** Node $X$ sees $R_{D1}$ in parallel with the source of $M_2$, $1/g_{m2}$ |
| **Answer 2** | **(a) First stage.** $\dfrac{v_X}{v_{in}}$ $=-g_{m1}\left(R_{D1}\parallel\dfrac{1}{g_{m2}}\right)$ $=-\dfrac{g_{m1}R_{D1}}{1+g_{m2}R_{D1}}$ |
| **Answer 3** | **(a) Second stage.** Common gate driven at its source by $v_X$: $\dfrac{v_{out}}{v_X}=g_{m2}R_{D2}$ |
| **Answer 4** | **(a) Overall.** $\dfrac{v_{out}}{v_{in}}$ $=-\dfrac{g_{m1}g_{m2}R_{D1}R_{D2}}{1+g_{m2}R_{D1}}$ |
| **Answer 5** | **(b) Limit.** $R_{D1}\to\infty$: $\dfrac{v_{out}}{v_{in}}$ $=-g_{m1}R_{D2}$ |
| **Answer 6** | **(b) Why.** All the small-signal drain current $g_{m1}v_{in}$ of $M_1$ must then flow through $M_2$ into $R_{D2}$: the pair is a cascode, $M_2$ only passes the current on |
| **Answer 7** | **Check.** Nodal analysis with $g_{m1}=3\,\mathrm{mS}$, $g_{m2}=5\,\mathrm{mS}$, $R_{D1}=2\,\mathrm{k}\Omega$, $R_{D2}=4\,\mathrm{k}\Omega$ gives the same number as the formula |
| **Result** | (a) $A_v$ $=-\dfrac{g_{m1}g_{m2}R_{D1}R_{D2}}{1+g_{m2}R_{D1}}$; (b) $A_v=-g_{m1}R_{D2}$ |

**Problem 133**

| | MOSFET |
|---|---|
| **Problem** | A PMOS transistor $M_1$ has its source at $V_{DD}$ and the input at its gate; its drain, node $X$, has a resistor $R_{D1}$ to ground, so $M_1$ and $R_{D1}$ form a common-source stage. Node $X$ also connects to the source of a common-gate NMOS $M_2$ (gate at $V_b$, drain resistor $R_{D2}$ to $V_{DD}$, output at the drain of $M_2$). With $\lambda=0$: (a) find $v_{out}/v_{in}$ $=(v_X/v_{in})(v_{out}/v_X)$; (b) simplify for $R_{D1}\to\infty$ and explain the result. |
| **Given** | $\lambda=0$; $g_{m1}$, $g_{m2}$, $R_{D1}$, $R_{D2}$ symbolic |
| **Answer 1** | **Read the circuit.** $M_1$ is a PMOS device with its source at $V_{DD}$ (ac ground) and its drain at $X$: a common-source device that delivers the signal current $-g_{m1}v_{in}$ into $X$ |
| **Answer 2** | **Impedance at X.** Node $X$ sees $R_{D1}$ to ground in parallel with the source of $M_2$: $R_{D1}\parallel\dfrac{1}{g_{m2}}$ $=\dfrac{R_{D1}}{1+g_{m2}R_{D1}}$ |
| **Answer 3** | **(a) First stage.** $\dfrac{v_X}{v_{in}}$ $=-g_{m1}\left(R_{D1}\parallel\dfrac{1}{g_{m2}}\right)$ $=-\dfrac{g_{m1}R_{D1}}{1+g_{m2}R_{D1}}$ |
| **Answer 4** | **(a) Second stage.** Common gate driven at its source by $v_X$: $\dfrac{v_{out}}{v_X}=g_{m2}R_{D2}$ |
| **Answer 5** | **(a) Overall.** $\dfrac{v_{out}}{v_{in}}$ $=-\dfrac{g_{m1}g_{m2}R_{D1}R_{D2}}{1+g_{m2}R_{D1}}$, the same expression as in Problem 132 |
| **Answer 6** | **(b) Limit.** $R_{D1}\to\infty$: $\dfrac{v_{out}}{v_{in}}$ $=-g_{m1}R_{D2}$ |
| **Answer 7** | **(b) Why.** With no signal current lost in $R_{D1}$, all of the drain signal current $g_{m1}v_{in}$ of $M_1$ enters the source of $M_2$ and flows on into $R_{D2}$: a folded cascode, in which $M_2$ only passes the current on |
| **Answer 8** | **Check.** Nodal analysis with $g_{m1}=3\,\mathrm{mS}$, $g_{m2}=5\,\mathrm{mS}$, $R_{D1}=2\,\mathrm{k}\Omega$, $R_{D2}=4\,\mathrm{k}\Omega$ gives $v_{out}/v_{in}=-10.9$, equal to the formula; for $R_{D1}\to\infty$ it gives $-12.0=-g_{m1}R_{D2}$ |
| **Result** | (a) $A_v$ $=-\dfrac{g_{m1}g_{m2}R_{D1}R_{D2}}{1+g_{m2}R_{D1}}$; (b) $A_v=-g_{m1}R_{D2}$ (folded cascode; same as Problem 132) |

**Problem 134**

| | MOSFET |
|---|---|
| **Problem** | NMOS $M_1$ is a source follower (input at the gate, drain at $V_{DD}$) whose source, node $X$, is biased by an ideal current source $I_1$ to ground and drives the source of a common-gate NMOS $M_2$ (gate at $V_b$, drain resistor $R_D$ to $V_{DD}$, output at the drain). With $\lambda=0$: (a) find $v_{out}/v_{in}$ $=(v_X/v_{in})(v_{out}/v_X)$; (b) simplify for $g_{m1}=g_{m2}$. |
| **Given** | $\lambda=0$; ideal $I_1$; $g_{m1}$, $g_{m2}$, $R_D$ symbolic |
| **Answer 1** | **Load of the follower.** At $X$ the ideal $I_1$ is an open circuit, so the only load is the source of $M_2$, $1/g_{m2}$ |
| **Answer 2** | **(a) First stage.** $\dfrac{v_X}{v_{in}}$ $=\dfrac{1/g_{m2}}{1/g_{m2}+1/g_{m1}}$ $=\dfrac{g_{m1}}{g_{m1}+g_{m2}}$ |
| **Answer 3** | **(a) Second stage.** $\dfrac{v_{out}}{v_X}=g_{m2}R_D$ |
| **Answer 4** | **(a) Overall.** $\dfrac{v_{out}}{v_{in}}$ $=\dfrac{g_{m1}g_{m2}}{g_{m1}+g_{m2}}R_D$ |
| **Answer 5** | **(b) Equal transconductances.** $g_{m1}=g_{m2}=g_m$: $\dfrac{v_{out}}{v_{in}}$ $=\dfrac{g_mR_D}{2}$, half of a common-source gain, as for a differential pair driven on one side |
| **Answer 6** | **Check.** Nodal analysis with $g_{m1}=3\,\mathrm{mS}$, $g_{m2}=5\,\mathrm{mS}$, $R_D=4\,\mathrm{k}\Omega$ matches; with $g_{m2}=g_{m1}$ it gives $g_mR_D/2$ |
| **Result** | (a) $A_v$ $=\dfrac{g_{m1}g_{m2}R_D}{g_{m1}+g_{m2}}$; (b) $A_v=g_mR_D/2$ |

**Problem 135**

| | MOSFET |
|---|---|
| **Problem** | Two identical NMOS common-source stages are cascaded (infinite input resistance each). Source resistance $R_{sig}=200\,\mathrm{k\Omega}$, $R_{D1}=R_{D2}=10\,\mathrm{k\Omega}$, load $R_L=10\,\mathrm{k\Omega}$ on the second drain; each stage is biased at $I_D=0.3\,\mathrm{mA}$, $V_{OV}=0.2\,\mathrm{V}$, $\lambda=0$. (a) Describe the small-signal circuit; (b) find $g_m$ and the overall gain $v_o/v_{sig}$. |
| **Given** | $R_{sig}=200\,\mathrm{k}\Omega$, $R_{D1}$ $=R_{D2}$ $=R_L$ $=10\,\mathrm{k}\Omega$, $I_D=0.3\,\mathrm{mA}$, $V_{OV}=0.2\,\mathrm{V}$ |
| **Answer 1** | **(a) Small-signal circuit.** $v_{sig}$ with $R_{sig}$ drives $v_{gs1}$; $g_{m1}v_{gs1}$ flows into $R_{D1}$, the only load of stage 1 (the gate of stage 2 draws no current); $v_{gs2}=v_{o1}$ and $g_{m2}v_{gs2}$ flows into $R_{D2}\parallel R_L$. |
| **Answer 2** | **(b) Transconductance.** $g_{m1}$ $=g_{m2}$ $=\dfrac{2I_D}{V_{OV}}$ $=\dfrac{0.6}{0.2}$ $=3\,\mathrm{mA/V}$. |
| **Answer 3** | **Stage 1.** No gate current, so no drop in $R_{sig}$: $\dfrac{v_{gs2}}{v_{gs1}}$ $=-g_{m1}R_{D1}$ $=-30$. |
| **Answer 4** | **Stage 2.** $\dfrac{v_o}{v_{gs2}}$ $=-g_{m2}(R_{D2}\parallel R_L)$ $=-3\times5$ $=-15$. |
| **Answer 5** | **Overall.** $G_v=(-30)(-15)=450\,\mathrm{V/V}$. |
| **Result** | $g_m=3\,\mathrm{mA/V}$, $G_v=450\,\mathrm{V/V}$ |

**Problem 136**

| | MOSFET |
|---|---|
| **Problem** | Two identical NMOS stages with $g_m=10\,\mathrm{mA/V}$ ($r_o$ neglected) are cascaded. Stage 1 is a source follower with a $10\,\mathrm{k\Omega}$ source resistor. Stage 2 is a common-gate stage whose source (input) has a $10\,\mathrm{k\Omega}$ bias resistor to ground and whose drain resistor is $5\,\mathrm{k\Omega}$ with a $2\,\mathrm{k\Omega}$ load. Find: (a) the open-circuit voltage gain $A_{vo}$ and output resistance $R_o$ of stage 1; (b) the input resistance $R_{in}$ of stage 2 and its gain $v_o/v_{i2}$; (c) the overall gain $v_o/v_i$ with stage 1 driving stage 2. |
| **Given** | $g_m=10\,\mathrm{mA/V}$, $R_{S1}=10\,\mathrm{k\Omega}$, $R_{S2}=10\,\mathrm{k\Omega}$, $R_D=5\,\mathrm{k\Omega}$, $R_L=2\,\mathrm{k\Omega}$ |
| **Answer 1** | **(a) Open-circuit gain.** $A_{vo}$ $=\dfrac{R_{S1}}{R_{S1}+1/g_m}$ $=\dfrac{10}{10+0.1}$ $=0.99\,\mathrm{V/V}$. |
| **Answer 2** | **(a) Output resistance.** $R_o$ $=\dfrac{1}{g_m}\parallel R_{S1}$ $=0.1\parallel10$ $=0.099\,\mathrm{k\Omega}$ $=99\,\Omega$. |
| **Answer 3** | **(b) Input resistance of stage 2.** $R_{in}$ $=R_{S2}\parallel\dfrac{1}{g_m}$ $=10\parallel0.1$ $=99\,\Omega$. |
| **Answer 4** | **(b) Gain of stage 2.** $\dfrac{v_o}{v_{i2}}$ $=g_m(R_D\parallel R_L)$ $=10\times(5\parallel2)$ $=14.3\,\mathrm{V/V}$. |
| **Answer 5** | **(c) Loading between stages.** $v_{i2}$ $=A_{vo}v_i\dfrac{R_{in}}{R_{in}+R_o}$ $=0.99\,v_i\dfrac{99}{99+99}$ $=0.495\,v_i$. |
| **Answer 6** | **(c) Overall gain.** $\dfrac{v_o}{v_i}$ $=14.3\times0.495$ $=7.07\,\mathrm{V/V}$. |
| **Result** | $A_{vo}=0.99$, $R_o=99\,\Omega$; $R_{in}=99\,\Omega$, $v_o/v_{i2}=14.3$; overall $\approx7.1\,\mathrm{V/V}$ |

**Problem 137**

| | MOSFET |
|---|---|
| **Problem** | Two common-source stages are cascaded; each has a current-source load of the opposite polarity, so the output resistance of each stage is $r_o/2$ (its own $r_o$ in parallel with the load's $r_o$). All transistors carry the same current, have the same $\lvert V_A\rvert=5\,\mathrm{V}$ and the same overdrive $\lvert V_{OV}\rvert$. Find the overall gain in terms of $\lvert V_{OV}\rvert$ and the overdrive needed for $A_v=400\,\mathrm{V/V}$. |
| **Given** | $\lvert V_A\rvert=5\,\mathrm{V}$ (all devices), equal $I_D$, equal $\lvert V_{OV}\rvert$, $A_v=400\,\mathrm{V/V}$ |
| **Answer 1** | **Stage gains.** $A_{v1}=-g_{m1}\dfrac{r_{o1}}2$ and $A_{v2}=-g_{m2}\dfrac{r_{o2}}2$. |
| **Answer 2** | **Equal parameters.** Equal currents and $\lvert V_A\rvert$ give $r_{o1}=r_{o2}=r_o$; equal $V_{OV}$ gives $g_{m1}=g_{m2}=g_m$. |
| **Answer 3** | **Overall gain.** $A_v$ $=A_{v1}A_{v2}$ $=\dfrac14(g_mr_o)^2$. |
| **Answer 4** | **Intrinsic gain.** $g_mr_o$ $=\dfrac{2\lvert V_A\rvert}{\lvert V_{OV}\rvert}$ $=\dfrac{10}{\lvert V_{OV}\rvert}$, so $A_v$ $=\dfrac14\left(\dfrac{10}{\lvert V_{OV}\rvert}\right)^2$. |
| **Answer 5** | **Solve.** $400$ $=\dfrac14\left(\dfrac{10}{\lvert V_{OV}\rvert}\right)^2\Rightarrow\dfrac{10}{\lvert V_{OV}\rvert}$ $=40\Rightarrow\lvert V_{OV}\rvert$ $=0.25\,\mathrm{V}$. |
| **Result** | $A_v$ $=\dfrac14\left(\dfrac{10}{\lvert V_{OV}\rvert}\right)^2$; $\lvert V_{OV}\rvert$ $=0.25\,\mathrm{V}$ |

**Problem 138**

| | MOSFET |
|---|---|
| **Problem** | A CMOS inverter-type amplifier has NMOS $Q_1$ and PMOS $Q_2$ with gates joined (input) and drains joined (output), supplies $\pm1\,\mathrm{V}$, and a resistor $R=1\,\mathrm{M}\Omega$ connected between input and output. The transistors are matched with $k_n'(W/L)$ $=k_p'(W/L)$ $=1\,\mathrm{mA/V^2}$, $\lvert V_t\rvert=0.5\,\mathrm{V}$ and $\lvert V_A\rvert=20\,\mathrm{V}$. (a) Find the dc bias current $I_D$ with the input/output left open. (b) Neglecting $r_o$, find the voltage gain $v_o/v_i$. (c) Including $r_o$, find $v_o/v_i$ and $R_{in}$. (d) With a source resistance $R_{sig}=20\,\mathrm{k}\Omega$, find $v_o/v_{sig}$. (e) Find the output voltage range for saturation. |
| **Given** | $V_{DD}=1\,\mathrm{V}$, $V_{SS}=-1\,\mathrm{V}$, $R=1\,\mathrm{M}\Omega$, $k_n'(W/L)$ $=k_p'(W/L)$ $=1\,\mathrm{mA/V^2}$, $\lvert V_t\rvert=0.5\,\mathrm{V}$, $\lvert V_A\rvert=20\,\mathrm{V}$, $R_{sig}=20\,\mathrm{k}\Omega$ |
| **Answer 1** | **(a) Bias.** No dc current flows in the gates or in $R$, so $V_D=V_G$ and, by symmetry, $V_{GS1}$ $=V_{SG2}$ $=\tfrac12[1-(-1)]$ $=1\,\mathrm{V}$ and $V_G=0$. $I_D$ $=\tfrac12(1)(1-0.5)^2$ $=0.125\,\mathrm{mA}$. |
| **Answer 2** | **(b) Model.** Both transistors have $v_{gs}=v_i$, so their signal currents add to $2g_mv_i$; $g_m$ $=\dfrac{2I_D}{V_{OV}}$ $=\dfrac{2\times0.125}{0.5}$ $=0.5\,\mathrm{mA/V}$. |
| **Answer 3** | **(b) Gain neglecting $r_o$.** All of $2g_mv_i$ flows through $R$: $v_o$ $=v_i-2g_mRv_i\Rightarrow A_v$ $=1-2g_mR$ $=1-2\times0.5\times1000$ $=-999\,\mathrm{V/V}$. |
| **Answer 4** | **(c) Gain with $r_o$.** $r_o$ $=\dfrac{20}{0.125}$ $=160\,\mathrm{k}\Omega$; node at the output: $\dfrac{v_o}{r_o/2}+2g_mv_i+\dfrac{v_o-v_i}{R}$ $=0\Rightarrow A_v$ $=\dfrac{1/R-2g_m}{2/r_o+1/R}$ $=-74.0\,\mathrm{V/V}$ (the approximation $-2g_m(R\parallel r_o/2)=-74.1$ gives the same). |
| **Answer 5** | **(c) Input resistance.** $i_i$ $=(v_i-v_o)/R\Rightarrow R_{in}$ $=\dfrac{R}{1-A_v}$ $=\dfrac{1000}{1+74.0}$ $=13.3\,\mathrm{k}\Omega$. |
| **Answer 6** | **(d) With source resistance.** $\dfrac{v_i}{v_{sig}}$ $=\dfrac{13.3}{13.3+20}$ $=0.4$, so $\dfrac{v_o}{v_{sig}}$ $=0.4\times(-74.0)$ $=-29.6\,\mathrm{V/V}$. |
| **Answer 7** | **(e) Output range.** Each transistor needs $\lvert V_{OV}\rvert$ $=0.5\,\mathrm{V}$ across it to stay saturated, so $-0.5\,\mathrm{V}\le v_O\le+0.5\,\mathrm{V}$. |
| **Result** | $I_D=0.125\,\mathrm{mA}$; $A_v=-999$ (no $r_o$), $-74.0$ (with $r_o$); $R_{in}=13.3\,\mathrm{k}\Omega$; $v_o/v_{sig}=-29.6$; $-0.5\,\mathrm{V}\le v_O\le0.5\,\mathrm{V}$ |
