**Problem 1**

| | BJT |
|---|---|
| **Problem** | An npn transistor $Q_1$ has $I_S=5\times10^{-16}\,\mathrm{A}$. Its emitter is grounded and a $750\,\mathrm{mV}$ battery is connected between base and ground. The collector is joined to a $3\,\mathrm{V}$ supply through $R_L=1\,\mathrm{k}\Omega$; $V_{out}$ is the collector voltage. Find $I_C$ and $V_{out}$ and confirm the active mode (Early effect ignored). |
| **Given** | $I_S=5\times10^{-16}\,\mathrm{A}$, $V_{BE}=750\,\mathrm{mV}$, $R_L=1\,\mathrm{k}\Omega$, $V_{CC}=3\,\mathrm{V}$, $V_T=26\,\mathrm{mV}$ |
| **Answer 1** | **Collector current.** $I_C$ $=I_S\exp\dfrac{V_{BE}}{V_T}$ $=5\times10^{-16}\,\mathrm{A}\times\exp\dfrac{0.75}{0.026}$ $=1.69\,\mathrm{mA}$. |
| **Answer 2** | **Drop on the load.** $R_LI_C$ $=1\,\mathrm{k}\Omega\times1.69\,\mathrm{mA}$ $=1.69\,\mathrm{V}$. |
| **Answer 3** | **Output voltage.** $V_{out}$ $=V_{CC}-R_LI_C$ $=3\,\mathrm{V}-1.69\,\mathrm{V}$ $=1.31\,\mathrm{V}$. |
| **Answer 4** | **Check the region.** $V_{out}$ $=1.31\,\mathrm{V}>V_{BE}$ $=0.75\,\mathrm{V}$, so the collector is above the base and the base-collector junction is reverse-biased: active mode, as assumed. |
| **Result** | $I_C=1.69\,\mathrm{mA}$, $V_{out}=1.31\,\mathrm{V}$ (active mode) |

**Problem 2**

| | BJT |
|---|---|
| **Problem** | An npn transistor with $I_S=6\times10^{-16}\,\mathrm{A}$ has its emitter grounded and a $1\,\mathrm{k}\Omega$ resistor from the collector to $V_{CC}$. A microphone, which produces no dc level, is connected directly between base and ground, so the base-emitter voltage equals the microphone signal. If the signal peaks at $20\,\mathrm{mV}$, what is the peak output signal at the collector? Use $V_T=26\,\mathrm{mV}$. |
| **Given** | $I_S=6\times10^{-16}\,\mathrm{A}$, $R_C=1\,\mathrm{k}\Omega$, $v_{BE,peak}=20\,\mathrm{mV}$, no dc bias, $V_T=26\,\mathrm{mV}$ |
| **Answer 1** | **Bias situation.** The microphone has no dc component, so $V_{BE}=0$ at rest and the bias collector current is essentially zero (about $I_S$); hence $g_m=I_C/V_T\approx0$. |
| **Answer 2** | **Collector current at the signal peak.** $I_C$ $=I_S\exp\dfrac{V_{BE}}{V_T}$ $=6\times10^{-16}\exp\dfrac{20}{26}$ $=1.29\times10^{-15}\,\mathrm{A}$. |
| **Answer 3** | **Output voltage.** $R_CI_C$ $=1\,\mathrm{k}\Omega\times1.29\times10^{-15}\,\mathrm{A}$ $=1.29\times10^{-12}\,\mathrm{V}$. |
| **Answer 4** | **More careful view.** Measured from the rest value $I_S$ the change is $I_S(e^{20/26}-1)$ $=6.95\times10^{-16}\,\mathrm{A}$, giving $0.695\times10^{-12}\,\mathrm{V}$: still of order a picovolt. |
| **Answer 5** | **Conclusion.** The output is about $10^{-12}\,\mathrm{V}$, i.e. no amplification at all, because with no bias the transconductance is zero. The transistor must be biased first. |
| **Result** | Peak output about $1.3\times10^{-12}\,\mathrm{V}$ ($I_C$ $\approx1.29\times10^{-15}\,\mathrm{A}$): essentially none |

**Problem 3**

| | BJT |
|---|---|
| **Problem** | An npn has its emitter grounded, its base held at $V_B=0.8\,\mathrm{V}$, and its collector (node $X$) connected to $V_{CC}=2.5\,\mathrm{V}$ through $R_C=1\,\mathrm{k}\Omega$; $I_S=5\times10^{-17}\,\mathrm{A}$. Find $V_X$ for (a) $V_A=\infty$ and (b) $V_A=5\,\mathrm{V}$. |
| **Given** | $V_B=0.8\,\mathrm{V}$, $R_C=1\,\mathrm{k}\Omega$, $V_{CC}=2.5\,\mathrm{V}$, $I_S=5\times10^{-17}\,\mathrm{A}$; $V_A=\infty$ or $5\,\mathrm{V}$ |
| **Answer 1** | **Base-emitter law.** $I_0$ $=I_S\exp(V_B/V_T)$ $=5\times10^{-17}\exp(0.8/0.026)$ $=1.15\,\mathrm{mA}$. |
| **Answer 2** | **(a) No Early effect.** $I_C=1.15\,\mathrm{mA}$, $V_X$ $=V_{CC}-R_CI_C$ $=2.5-1.153$ $=1.35\,\mathrm{V}$ (active, $V_X>V_B$). |
| **Answer 3** | **(b) Two equations.** $I_C=I_0(1+V_X/V_A)$ and $V_X=V_{CC}-R_CI_C$. |
| **Answer 4** | **(b) Solve for VX.** $V_X$ $=\dfrac{V_{CC}-R_CI_0}{1+R_CI_0/V_A}$ $=\dfrac{2.5-1.153}{1+1.153/5}$ $=1.09\,\mathrm{V}$. |
| **Answer 5** | **(b) Current.** $I_C$ $=1.153\,\mathrm{mA}\,(1+1.094/5)$ $=1.41\,\mathrm{mA}$; still active ($V_X>V_B$). |
| **Result** | (a) $V_X=1.35\,\mathrm{V}$; (b) $V_X=1.09\,\mathrm{V}$ ($I_C=1.41\,\mathrm{mA}$) |

**Problem 1**

| | MOSFET |
|---|---|
| **Problem** | An NMOS transistor with $V_{tn}=0.5\,\mathrm{V}$ and $k_n$ $=k_n^\prime W/L$ $=2\,\mathrm{mA/V^2}$ has its source grounded and its gate at $2\,\mathrm{V}$. Its drain is connected to a $2\,\mathrm{V}$ supply through $R_D=13.1\,\mathrm{k}\Omega$. Find $V_{DS}$, $I_D$ and the region of operation. |
| **Given** | $V_{DD}=2\,\mathrm{V}$, $V_{GS}=2\,\mathrm{V}$, $V_{tn}=0.5\,\mathrm{V}$, $k_n=2\,\mathrm{mA/V^2}$, $R_D=13.1\,\mathrm{k}\Omega$ |
| **Answer 1** | **Assume triode.** $V_{OV}=2-0.5=1.5\,\mathrm{V}$; $I_D$ $=k_n\!\left(V_{OV}V_{DS}-\tfrac12V_{DS}^2\right)$ and $I_D=\dfrac{2-V_{DS}}{13.1}$ |
| **Answer 2** | **Equate.** $\dfrac{2-V_{DS}}{13.1}$ $=2\!\left(1.5V_{DS}-\tfrac12V_{DS}^2\right)\Rightarrow V_{DS}^2-3.076V_{DS}+0.153$ $=0$ |
| **Answer 3** | **Solve.** $V_{DS}$ $=\dfrac{3.076-\sqrt{3.076^2-4\times0.153}}{2}$ $=0.0505\,\mathrm{V}$ $\approx0.05\,\mathrm{V}$ (the other root, $3.03\,\mathrm{V}$, exceeds $V_{DD}$) |
| **Answer 4** | **Check region.** $V_{DS}$ $=0.05\,\mathrm{V}<V_{OV}$ $=1.5\,\mathrm{V}$: triode, as assumed |
| **Answer 5** | **Drain current.** $I_D$ $=\dfrac{2-0.0505}{13.1\,\mathrm{k}\Omega}$ $=0.149\,\mathrm{mA}$ $\approx0.15\,\mathrm{mA}$ |
| **Result** | $V_{DS}\approx0.05\,\mathrm{V}$ (triode), $I_D\approx0.15\,\mathrm{mA}$ |

**Problem 4**

| | BJT |
|---|---|
| **Problem** | To fix the bias problem of the previous amplifier (npn, emitter grounded, $R_C=1\,\mathrm{k}\Omega$ to $V_{CC}=2.5\,\mathrm{V}$, microphone between base and ground), a student wires the base node directly to $V_{CC}$ so that the base-emitter junction is forward biased. Explain why this does not work. |
| **Given** | $V_{CC}=2.5\,\mathrm{V}$, $R_C=1\,\mathrm{k}\Omega$, base connected directly to $V_{CC}$ |
| **Answer 1** | **Signal path.** The base node is connected to an ideal voltage source $V_{CC}$, which keeps it at exactly $2.5\,\mathrm{V}$ whatever the microphone does: the microphone signal is shorted by the supply. |
| **Answer 2** | **No signal at the output.** Because $V_{BE}$ stays constant, $I_C$ and hence $V_{out}=V_{CC}-R_CI_C$ stay constant. The gain is zero. |
| **Answer 3** | **Bias current.** Also $V_{BE}=V_{CC}-0=2.5\,\mathrm{V}$. With $I_C=I_S\exp(V_{BE}/V_T)$ and e.g. $I_S=6\times10^{-16}\,\mathrm{A}$, the exponent is $96.2$ and the formula gives $3\times10^{26}\,\mathrm{A}$; the real current would be limited only by parasitic resistances and would destroy the device. |
| **Answer 4** | **What is needed.** The base must be connected to the supply through a (large) resistance that sets a moderate base current, or through a divider, so that $V_{BE}$ is about 0.7 to 0.8 V and the signal can still move the base voltage. |
| **Result** | The supply short-circuits the signal ($V_{BE}$ constant, no amplification) and forces an enormous bias current ($V_{BE}=2.5\,\mathrm{V}$) |

**Problem 5**

| | BJT |
|---|---|
| **Problem** | An npn transistor ($\beta=100$, $I_S=10^{-17}\,\mathrm{A}$) has its emitter grounded, a $100\,\mathrm{k}\Omega$ resistor $R_B$ from $V_{CC}=2.5\,\mathrm{V}$ to the base and a $1\,\mathrm{k}\Omega$ resistor $R_C$ from $V_{CC}$ to the collector. Find the collector bias current by iterating on $V_{BE}$ (start from $0.8\,\mathrm{V}$) and verify that the transistor is in the forward active region. Use $V_T=26\,\mathrm{mV}$. |
| **Given** | $V_{CC}=2.5\,\mathrm{V}$, $R_B=100\,\mathrm{k}\Omega$, $R_C=1\,\mathrm{k}\Omega$, $\beta=100$, $I_S=10^{-17}\,\mathrm{A}$, $V_T=26\,\mathrm{mV}$ |
| **Answer 1** | **Base loop.** $R_BI_B+V_{BE}=V_{CC}$, so $I_B=\dfrac{V_{CC}-V_{BE}}{R_B}$ and $I_C=\beta I_B$. |
| **Answer 2** | **First guess $V_{BE}=0.8\,\mathrm{V}$.** $I_B$ $=\dfrac{2.5-0.8}{100\,\mathrm{k}\Omega}$ $=17.0\,\mu\mathrm{A}$, $I_C=1.70\,\mathrm{mA}$. |
| **Answer 3** | **Update $V_{BE}$.** $V_{BE}$ $=V_T\ln\dfrac{I_C}{I_S}$ $=0.026\ln\dfrac{1.70\times10^{-3}}{10^{-17}}$ $=0.852\,\mathrm{V}$. |
| **Answer 4** | **Second pass.** $I_B$ $=\dfrac{2.5-0.852}{100\,\mathrm{k}\Omega}$ $=16.5\,\mu\mathrm{A}$, $I_C=1.65\,\mathrm{mA}$ (it changes by only 3 %; the next $V_{BE}$ would be $0.851\,\mathrm{V}$), so we stop. Iterating to convergence gives $I_C=1.649\,\mathrm{mA}$. |
| **Answer 5** | **Collector voltage.** $V_{CE}$ $=V_{CC}-R_CI_C$ $=2.5-1\,\mathrm{k}\Omega\times1.65\,\mathrm{mA}$ $=0.85\,\mathrm{V}$. |
| **Answer 6** | **Region.** $V_{CE}$ $\approx V_{BE}$ $=0.85\,\mathrm{V}$, so the base-collector voltage is about $0$: the transistor is just at the edge between active mode and saturation (still active, with no margin). |
| **Result** | $I_C\approx1.65\,\mathrm{mA}$, $V_{BE}\approx0.852\,\mathrm{V}$, $V_{CE}\approx0.85\,\mathrm{V}$: edge of active mode |

**Problem 6**

| | BJT |
|---|---|
| **Problem** | An npn transistor ($\beta=100$, $I_S=10^{-17}\,\mathrm{A}$, emitter grounded) is biased from $V_{CC}=2.5\,\mathrm{V}$ by a divider: $R_1=17\,\mathrm{k}\Omega$ from $V_{CC}$ to the base and $R_2=8\,\mathrm{k}\Omega$ from the base to ground; $R_C=5\,\mathrm{k}\Omega$ connects the collector to $V_{CC}$. Neglecting the base current in the divider, find $I_C$, check that $I_B$ is negligible and that the transistor is active. Use $V_T=26\,\mathrm{mV}$. |
| **Given** | $V_{CC}=2.5\,\mathrm{V}$, $R_1=17\,\mathrm{k}\Omega$, $R_2=8\,\mathrm{k}\Omega$, $R_C=5\,\mathrm{k}\Omega$, $\beta=100$, $I_S=10^{-17}\,\mathrm{A}$ |
| **Answer 1** | **Base voltage.** Neglecting $I_B$: $V_X$ $=\dfrac{R_2}{R_1+R_2}V_{CC}$ $=\dfrac{8}{25}\times2.5$ $=0.800\,\mathrm{V}$, so $V_{BE}=0.800\,\mathrm{V}$. |
| **Answer 2** | **Collector current.** $I_C$ $=I_S\exp\dfrac{V_{BE}}{V_T}$ $=10^{-17}\exp\dfrac{0.8}{0.026}$ $=231\,\mu\mathrm{A}$. |
| **Answer 3** | **Base current.** $I_B=I_C/\beta=2.31\,\mu\mathrm{A}$. |
| **Answer 4** | **Is $I_B$ negligible?.** It must be small against the divider current $\dfrac{V_{CC}}{R_1+R_2}$ $=\dfrac{2.5}{25\,\mathrm{k}\Omega}$ $=100\,\mu\mathrm{A}$. Indeed $100\,\mu\mathrm{A}\approx43I_B$. |
| **Answer 5** | **Collector voltage.** $V_{CE}$ $=V_{CC}-R_CI_C$ $=2.5-5\,\mathrm{k}\Omega\times231\,\mu\mathrm{A}$ $=1.35\,\mathrm{V}$. |
| **Answer 6** | **Region.** $V_{CE}$ $=1.35\,\mathrm{V}>V_{BE}$ $=0.80\,\mathrm{V}$: the base-collector junction is reverse biased, the transistor is in the forward active region. |
| **Result** | $V_X=0.800\,\mathrm{V}$, $I_C=231\,\mu\mathrm{A}$, $I_B=2.31\,\mu\mathrm{A}$ ($\ll100\,\mu\mathrm{A}$), $V_{CE}=1.35\,\mathrm{V}$, active |

**Problem 7**

| | BJT |
|---|---|
| **Problem** | An npn transistor ($\beta=100$, $I_S=10^{-17}\,\mathrm{A}$, emitter grounded) is biased from $V_{CC}=2.5\,\mathrm{V}$ by a divider $R_1=170\,\mathrm{k}\Omega$ (to $V_{CC}$) and $R_2=80\,\mathrm{k}\Omega$ (to ground) at the base; the collector resistor is $R_C=5\,\mathrm{k}\Omega$. This time the base current is not negligible. Find the collector bias current by replacing the divider with its Thevenin equivalent and iterating on $V_{BE}$. Use $V_T=26\,\mathrm{mV}$. |
| **Given** | $V_{CC}=2.5\,\mathrm{V}$, $R_1=170\,\mathrm{k}\Omega$, $R_2=80\,\mathrm{k}\Omega$, $R_C=5\,\mathrm{k}\Omega$, $\beta=100$, $I_S=10^{-17}\,\mathrm{A}$ |
| **Answer 1** | **Thevenin equivalent.** $V_{Thev}$ $=\dfrac{R_2}{R_1+R_2}V_{CC}$ $=0.800\,\mathrm{V}$, $R_{Thev}$ $=R_1\parallel R_2$ $=54.4\,\mathrm{k}\Omega$. |
| **Answer 2** | **Equations.** $V_{BE}=V_{Thev}-I_BR_{Thev}$, $I_C=\beta I_B$ and $V_{BE}=V_T\ln(I_C/I_S)$. Iterate in the order $V_{BE}\to I_B\to I_C\to V_{BE}$. |
| **Answer 3** | **First pass, guess $V_{BE}=0.750\,\mathrm{V}$.** $I_B$ $=\dfrac{0.800-0.750}{54.4\,\mathrm{k}\Omega}$ $=0.919\,\mu\mathrm{A}$, $I_C=91.9\,\mu\mathrm{A}$, new $V_{BE}=0.776\,\mathrm{V}$. |
| **Answer 4** | **Second and third pass.** $V_{BE}=0.776\,\mathrm{V}$: $I_B=0.440\,\mu\mathrm{A}$, $I_C=44.0\,\mu\mathrm{A}$, new $V_{BE}=0.757\,\mathrm{V}$; then $I_B=0.792\,\mu\mathrm{A}$, $I_C=79.2\,\mu\mathrm{A}$, new $V_{BE}=0.772\,\mathrm{V}$. The iteration swings around the answer because the exponential is so steep. |
| **Answer 5** | **Converged solution.** Solving $I_S e^{V_{BE}/V_T}$ $=\beta(V_{Thev}-V_{BE})/R_{Thev}$ directly (bisection) gives $V_{BE}=0.766\,\mathrm{V}$, $I_C=62.4\,\mu\mathrm{A}$, $I_B=0.624\,\mu\mathrm{A}$. |
| **Answer 6** | **Check the region.** $V_Y$ $=V_{CC}-R_CI_C$ $=2.5-5\,\mathrm{k}\Omega\times62.4\,\mu\mathrm{A}$ $=2.19\,\mathrm{V}>V_{BE}$: active. |
| **Answer 7** | **Why the Thevenin form matters.** Neglecting $I_B$ would give $V_{BE}=0.800\,\mathrm{V}$ and $I_C=231\,\mu\mathrm{A}$, nearly four times too large. |
| **Result** | $V_{BE}\approx0.766\,\mathrm{V}$, $I_C\approx62\,\mu\mathrm{A}$ (published: $63\,\mu\mathrm{A}$) |

**Problem 8**

| | BJT |
|---|---|
| **Problem** | An npn transistor $Q_1$ with grounded emitter has its base fed from a voltage source $V_B$ through a series resistor $R_1=10\,\mathrm{k}\Omega$. With $\beta=100$ and $I_S=7\times10^{-16}\,\mathrm{A}$, find $V_B$ for $I_C=1\,\mathrm{mA}$. |
| **Given** | $R_1=10\,\mathrm{k}\Omega$, $\beta=100$, $I_S=7\times10^{-16}\,\mathrm{A}$, $I_C=1\,\mathrm{mA}$, $V_T=26\,\mathrm{mV}$ |
| **Answer 1** | **Base-emitter voltage.** $V_{BE}$ $=V_T\ln\dfrac{I_C}{I_S}$ $=26\,\mathrm{mV}\cdot\ln\dfrac{10^{-3}}{7\times10^{-16}}$ $=728\,\mathrm{mV}$. |
| **Answer 2** | **Base current.** $I_B=I_C/\beta=10\,\mu\mathrm{A}$, dropping $R_1I_B=0.1\,\mathrm{V}$. |
| **Answer 3** | **Base loop.** $V_B$ $=R_1I_B+V_{BE}$ $=0.100+0.728$ $=0.828\,\mathrm{V}$. |
| **Result** | $V_B=828\,\mathrm{mV}$ |

**Problem 10**

Find $V_B$ for $I_X=1\,\mathrm{mA}$. |

Find $I_X$ and $I_Y$. |

**Problem 12**

| | BJT |
|---|---|
| **Problem** | Find the bias point of three npn circuits, each with $V_{CC}=2.5\,\mathrm{V}$ and $100\,\mathrm{k}\Omega$ from $V_{CC}$ to the base of $Q_1$. (a) Collector resistor $500\,\Omega$, emitter grounded. (b) Collector resistor $1\,\mathrm{k}\Omega$; the emitter goes to a diode-connected $Q_2$ (identical) to ground. (c) Collector resistor $1\,\mathrm{k}\Omega$; the emitter is held at $+0.5\,\mathrm{V}$ by a battery. $\beta=100$, $I_S=6\times10^{-16}\,\mathrm{A}$, $V_A=\infty$. |
| **Given** | $V_{CC}=2.5\,\mathrm{V}$, $R_B=100\,\mathrm{k}\Omega$, $R_C=500\,\Omega$ (a), $1\,\mathrm{k}\Omega$ (b, c), $\beta=100$, $I_S=6\times10^{-16}\,\mathrm{A}$, $V_A=\infty$, $V_T=26\,\mathrm{mV}$ |
| **Answer 1** | **(a) Base loop.** $I_B=\dfrac{V_{CC}-V_{BE}}{R_B}$, $I_C=\beta I_B$, $V_{BE}=V_T\ln(I_C/I_S)$; iterate. |

**Problem 16**

| | BJT |
|---|---|
| **Problem** | An npn current source $Q_1$ (emitter grounded) has its base fed by a divider $R_1$ (from $V_{CC}$) and $R_2$ (to ground), so that $\dfrac{R_2}{R_1+R_2}V_{CC}$ $=V_T\ln\dfrac{I_1}{I_S}$ (base current neglected). Find the sensitivity $\partial I_1/\partial V_{CC}$ and explain why it is proportional to the transconductance of $Q_1$. |
| **Given** | $I_1$ $=I_S\exp\!\left(\dfrac{R_2V_{CC}}{(R_1+R_2)V_T}\right)$ |
| **Answer 1** | **Current as a function of the supply.** $I_1$ $=I_S\exp\!\left(\dfrac{R_2}{R_1+R_2}\cdot\dfrac{V_{CC}}{V_T}\right)$ |
| **Answer 2** | **Differentiate.** $\dfrac{\partial I_1}{\partial V_{CC}}$ $=\dfrac{R_2}{R_1+R_2}\cdot\dfrac{1}{V_T}I_1$ |
| **Answer 3** | **In terms of $g_m$.** $\dfrac{\partial I_1}{\partial V_{CC}}$ $=g_{m1}\dfrac{R_2}{R_1+R_2}$, with $g_{m1}=I_1/V_T$. |
| **Answer 4** | **Intuition.** A supply change $\Delta V_{CC}$ reaches the base through the divider as $\Delta V_{BE}$ $=\dfrac{R_2}{R_1+R_2}\Delta V_{CC}$; the transistor turns a base-emitter voltage change into a collector current change through its transconductance, $\Delta I_1=g_{m1}\Delta V_{BE}$. Supply noise is amplified as if it were an input signal. |
| **Answer 5** | **Numerical check.** For $I_S=6\times10^{-17}\,\mathrm{A}$, $V_{CC}=2.5\,\mathrm{V}$ and $R_2/(R_1+R_2)$ set for $I_1=1\,\mathrm{mA}$ (ratio 0.317): $g_{m1}R_2/(R_1+R_2)$ $=12.2\,\mathrm{mA/V}$, and a finite difference of the exponential gives the same; a 1 % change of $V_{CC}$ (25 mV) changes $I_1$ by about 36 %. |
| **Result** | $\dfrac{\partial I_1}{\partial V_{CC}}$ $=\dfrac{R_2}{R_1+R_2}\cdot\dfrac{I_1}{V_T}$ $=g_{m1}\dfrac{R_2}{R_1+R_2}$ |

**Problem 3**

| | MOSFET |
|---|---|
| **Problem** | An NMOS current source $M_1$ (source grounded) has its gate fed by a divider $R_1$ (from $V_{DD}$) and $R_2$ (to ground), so $I_1$ $=\dfrac12\mu_nC_{ox}\dfrac WL\left(\dfrac{R_2}{R_1+R_2}V_{DD}-V_{TH}\right)^2$. Find $\partial I_1/\partial V_{DD}$ and relate it to the transconductance of $M_1$. |
| **Given** | $I_1$ $=\tfrac12\mu_nC_{ox}(W/L)(V_{GS}-V_{TH})^2$, $V_{GS}=\dfrac{R_2}{R_1+R_2}V_{DD}$ |
| **Answer 1** | **Differentiate.** $\dfrac{\partial I_1}{\partial V_{DD}}$ $=\mu_nC_{ox}\dfrac WL\left(\dfrac{R_2}{R_1+R_2}V_{DD}-V_{TH}\right)\dfrac{R_2}{R_1+R_2}$ |
| **Answer 2** | **In terms of $g_m$.** Since $g_{m1}$ $=\mu_nC_{ox}\dfrac WL(V_{GS}-V_{TH})$: $\dfrac{\partial I_1}{\partial V_{DD}}$ $=g_{m1}\dfrac{R_2}{R_1+R_2}$ |
| **Answer 3** | **Intuition.** The divider passes the fraction $R_2/(R_1+R_2)$ of any supply change to the gate, and $M_1$ converts that gate-source change into drain current through $g_{m1}$. |
| **Answer 4** | **Numerical check.** With $\mu_nC_{ox}=100\,\mu\mathrm{A/V^2}$, $W/L=10$, $V_{TH}=0.4\,\mathrm{V}$, $V_{DD}=1.8\,\mathrm{V}$, $R_2/(R_1+R_2)=0.5$: $I_1=125\,\mu\mathrm{A}$, $g_{m1}=0.5\,\mathrm{mS}$, sensitivity $0.25\,\mathrm{mA/V}$, equal to a finite difference. |
| **Result** | $\dfrac{\partial I_1}{\partial V_{DD}}$ $=\mu_nC_{ox}\dfrac WL\left(\dfrac{R_2V_{DD}}{R_1+R_2}-V_{TH}\right)\dfrac{R_2}{R_1+R_2}$ $=g_{m1}\dfrac{R_2}{R_1+R_2}$ |

**Problem 4**

| | MOSFET |
|---|---|
| **Problem** | For the divider-biased NMOS current source of Problem 3, $I_1$ $=\dfrac12\mu_nC_{ox}\dfrac WL\left(\dfrac{R_2}{R_1+R_2}V_{DD}-V_{TH}\right)^2$, the threshold voltage varies from batch to batch. Find the sensitivity of $I_1$ to $V_{TH}$ and explain why the problem is worse at low supply voltages. |
| **Given** | $I_1$ $=\tfrac12\mu_nC_{ox}(W/L)(V_{GS}-V_{TH})^2$ with $V_{GS}$ fixed by the divider |
| **Answer 1** | **Differentiate.** $\dfrac{\partial I_1}{\partial V_{TH}}$ $=-\mu_nC_{ox}\dfrac WL\left(\dfrac{R_2}{R_1+R_2}V_{DD}-V_{TH}\right)$ $=-g_{m1}$ |
| **Answer 2** | **Relative sensitivity.** $\dfrac{\Delta I_1}{I_1}$ $=-\dfrac{2\,\Delta V_{TH}}{V_{GS}-V_{TH}}$ |
| **Answer 3** | **Low supply.** With a low $V_{DD}$ the gate voltage, and so the overdrive $V_{GS}-V_{TH}$, must be small (a few hundred millivolts at most, to leave headroom). The same $\Delta V_{TH}$ is then a larger fraction of the overdrive, so the relative error of $I_1$ grows as $1/(V_{GS}-V_{TH})$. |
| **Answer 4** | **Numerical check.** A 20 mV threshold shift gives $\Delta I_1/I_1$ $=-2\times0.02/0.5$ $=-8\,\%$ with a 0.5 V overdrive, but $-20\,\%$ with a 0.2 V overdrive (exact: $-7.8\,\%$ and $-19\,\%$). |
| **Result** | $\dfrac{\partial I_1}{\partial V_{TH}}$ $=-\mu_nC_{ox}\dfrac WL(V_{GS}-V_{TH})$ $=-g_{m1}$; $\dfrac{\Delta I_1}{I_1}$ $\approx-\dfrac{2\Delta V_{TH}}{V_{GS}-V_{TH}}$, large when the overdrive is small |

**Problem 10**

| | MOSFET |
|---|---|
| **Problem** | (a) For an NMOS transistor biased at fixed $V_{GS}$, show that the sensitivity of $I_D$ to $V_t$ is $S_{V_t}^{I_D}$ $=-\dfrac{2V_t}{V_{OV}}$, and find the change of $I_D$ for $V_t=0.5\,\mathrm{V}\pm5\%$ with $V_{OV}=0.25\,\mathrm{V}$. (b) With a fixed gate voltage $V_G$ and source resistor $R_S$, show that $S_{V_t}^{I_D}$ $=-\dfrac{2V_t}{V_{OV}+2I_DR_S}$, and find $R_S$ and $V_G$ that limit the change of $I_D$ to $\mp5\%$ at $I_D=0.1\,\mathrm{mA}$. |
| **Given** | $V_t=0.5\,\mathrm{V}\pm5\%$, $V_{OV}=0.25\,\mathrm{V}$, $I_D=0.1\,\mathrm{mA}$ |
| **Answer 1** | **(a) Derivative.** $I_D$ $=\tfrac12k_n(V_{GS}-V_t)^2\Rightarrow\dfrac{\partial I_D}{\partial V_t}$ $=-k_n(V_{GS}-V_t)$. |
| **Answer 2** | **(a) Sensitivity.** $S$ $=\dfrac{\partial I_D}{\partial V_t}\dfrac{V_t}{I_D}$ $=-\dfrac{k_nV_{OV}V_t}{\tfrac12k_nV_{OV}^2}$ $=-\dfrac{2V_t}{V_{OV}}$. |
| **Answer 3** | **(a) Numbers.** $S=-\dfrac{2\times0.5}{0.25}=-4$, so $\Delta I_D/I_D$ $=-4\times(\pm5\%)$ $=\mp20\%$. |
| **Answer 4** | **(b) Loop.** $V_G=V_{GS}+I_DR_S$, $V_{GS}=V_t+\sqrt{2I_D/k_n}$. Differentiate with respect to $V_t$: $1+\dfrac{1}{2\sqrt{2I_D/k_n}}\dfrac{2}{k_n}\dfrac{\partial I_D}{\partial V_t}+R_S\dfrac{\partial I_D}{\partial V_t}$ $=0$. |
| **Answer 5** | **(b) Solve.** $\dfrac{\partial I_D}{\partial V_t}$ $=-\dfrac{1}{1/\sqrt{2k_nI_D}+R_S}$; multiply by $V_t/I_D$ and use $V_{OV}=\sqrt{2I_D/k_n}$ (so $\dfrac{I_D}{\sqrt{2k_nI_D}}$ $=\dfrac{V_{OV}}{2}$): $S=-\dfrac{2V_t}{V_{OV}+2I_DR_S}$. |
| **Answer 6** | **(b) Requirement.** A $\pm5\%$ change of $V_t$ must give $\mp5\%$ in $I_D$, so $\lvert S\rvert=1$: $1$ $=\dfrac{2\times0.5}{0.25+2I_DR_S}\Rightarrow I_DR_S$ $=0.375\,\mathrm{V}$. |
| **Answer 7** | **(b) Values.** $R_S$ $=\dfrac{0.375\,\mathrm{V}}{0.1\,\mathrm{mA}}$ $=3.75\,\mathrm{k}\Omega$; $V_G$ $=V_t+V_{OV}+I_DR_S$ $=0.5+0.25+0.375$ $=1.125\,\mathrm{V}$. |

| **Answer 2** | **Split of the emitter current.** $I_E=I_B+I_C=I_C(1+1/\beta)$, so $I_C$ $=\dfrac{I_E}{1+1/\beta}$ $=\dfrac{1}{1+1/10}$ $=0.909\,\mathrm{mA}$. |
| **Answer 3** | **Base current.** $I_B=I_E-I_C=0.0909\,\mathrm{mA}$ (flowing out of the base to ground). |
| **Answer 4** | **Base diode.** $I_{SB}$ $=I_S/\beta$ $=10^{-16}\,\mathrm{A}$, $V_{EB}$ $=V_T\ln\dfrac{I_B}{I_{SB}}$ $=0.025\ln\dfrac{0.0909\times10^{-3}}{10^{-16}}$ $=0.688\,\mathrm{V}$. |
| **Answer 5** | **Emitter voltage.** $V_E=V_B+V_{EB}=0.688\,\mathrm{V}$. |
| **Answer 6** | **Substitute $\beta=1000$.** $I_C$ $=\dfrac{1}{1+1/1000}$ $=0.999\,\mathrm{mA}$. |
| **Answer 7** | **Change.** $\Delta I_C$ $=0.999-0.909$ $=0.090\,\mathrm{mA}$, an increase of $9.9\,\%$. |
| **Result** | $I_C=0.909\,\mathrm{mA}$, $I_B=0.0909\,\mathrm{mA}$, $V_{EB}=0.688\,\mathrm{V}=V_E$; with $\beta=1000$, $I_C=0.999\,\mathrm{mA}$ ($+9.9\,\%$) |

| **Answer 1** | **Base voltage and current.** $V_B=3-0.7=2.3\,\mathrm{V}$, $I_B$ $=\dfrac{2.3}{100\,\mathrm{k\Omega}}$ $=0.023\,\mathrm{mA}$. |
| **Answer 2** | **Mode for $V_C=2\,\mathrm{V}$.** $V_C$ $=2\,\mathrm{V}<V_B$ $=2.3\,\mathrm{V}$ (collector junction reverse biased): active. |

| **Answer 5** | **$\beta=100$, active guess.** $I_C$ $=100\times0.023$ $=2.3\,\mathrm{mA}$, which would give $V_C=2.3\times1.74=4\,\mathrm{V}$: above the emitter and above the base, impossible. |

**Problem 58**

| | MOSFET |
|---|---|
| **Problem** | An NMOS transistor with its gate grounded through $R_G$, a source resistor $R_S$ to $-V_{SS}$ and a drain resistor, has $I_D=K(V_{GS}-V_t)^2$ with $K=\tfrac12k_n(W/L)$. (a) Show that the sensitivity of $I_D$ to $K$ is $S_K^{I_D}$ $=\dfrac{K}{I_D}\dfrac{\partial I_D}{\partial K}$ $=\dfrac{1}{1+2\sqrt{KI_D}\,R_S}$. (b) With $K=100\,\mu\mathrm{A/V^2}\pm10\%$, $V_t=1\,\mathrm{V}$ and $I_D=100\,\mu\mathrm{A}$ required to within $\pm1\%$, find $R_S$ and $V_{SS}$. (c) Find the variation of $I_D$ if $V_{SS}=5\,\mathrm{V}$ is used instead, with $V_{GS}$ unchanged. |
| **Given** | $K=100\,\mu\mathrm{A/V^2}\pm10\%$, $V_t=1\,\mathrm{V}$, $I_D=100\,\mu\mathrm{A}\pm1\%$, $V_G=0$ |
| **Answer 1** | **(a) Loop equation.** $V_{GS}+I_DR_S=V_{SS}$ with $V_{GS}=V_t+\sqrt{I_D/K}$, so $V_t+\sqrt{I_D/K}+I_DR_S=V_{SS}$. |
| **Answer 2** | **(a) Differentiate.** $\dfrac{\partial}{\partial K}$: $\dfrac{1}{2\sqrt{I_D/K}}\left(\dfrac{1}{K}\dfrac{\partial I_D}{\partial K}-\dfrac{I_D}{K^2}\right)+R_S\dfrac{\partial I_D}{\partial K}$ $=0$. |
| **Answer 3** | **(a) Solve.** $\dfrac{\partial I_D}{\partial K}\left[\dfrac{1}{2\sqrt{KI_D}}+R_S\right]$ $=\dfrac{1}{2}\dfrac{\sqrt{I_D}}{K^{3/2}}\Rightarrow S_K^{I_D}$ $=\dfrac{K}{I_D}\dfrac{\partial I_D}{\partial K}$ $=\dfrac{1}{1+2\sqrt{KI_D}R_S}$. |
| **Answer 4** | **(b) Required sensitivity.** $S$ $=\dfrac{\Delta I_D/I_D}{\Delta K/K}$ $=\dfrac{0.01}{0.10}$ $=0.1$. |
| **Answer 5** | **(b) $R_S$.** $\sqrt{KI_D}$ $=\sqrt{0.1\times0.1}$ $=0.1\,\mathrm{mA/V}$ (units mA, V). $0.1$ $=\dfrac{1}{1+0.2R_S}\Rightarrow R_S$ $=45\,\mathrm{k}\Omega$. |
| **Answer 6** | **(b) $V_{GS}$ and $V_{SS}$.** $100$ $=100(V_{GS}-1)^2\Rightarrow V_{GS}$ $=2\,\mathrm{V}$; $V_{SS}$ $=V_{GS}+I_DR_S$ $=2+0.1\times45$ $=6.5\,\mathrm{V}$. |
| **Answer 7** | **(c) $V_{SS}=5\,\mathrm{V}$.** $I_DR_S$ $=5-2$ $=3\,\mathrm{V}\Rightarrow R_S$ $=30\,\mathrm{k}\Omega$; $S$ $=\dfrac{1}{1+0.2\times30}$ $=\dfrac17$. |
| **Answer 8** | **(c) Variation.** $\dfrac{\Delta I_D}{I_D}$ $=\dfrac17\times(\pm10\%)$ $=\pm1.43\%$. |
| **Result** | (b) $R_S=45\,\mathrm{k}\Omega$, $V_{SS}=6.5\,\mathrm{V}$; (c) $R_S=30\,\mathrm{k}\Omega$, $\Delta I_D/I_D=\pm1.4\%$ |
