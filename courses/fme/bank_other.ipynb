| **Given** | $V_{CC}$ with ripple $v_r$ (frequency $2f_\mathrm{line}$); $V_{out}=V_{CC}-R_CI_C$ |

| **Answer 3** | **Effect of the ripple.** If $V_{CC}$ moves by $v_r$, then $V_{out}$ moves by $v_r$ as well: $\Delta V_{out}/\Delta V_{CC}=1$. The output therefore follows the supply with unity gain. |
| **Answer 4** | **Two components at the output.** $v_{out}=A_vv_{in}+v_r$: the amplified microphone signal plus the whole ripple. |
| **Answer 5** | **Source of the hum.** The ripple is a steady component at $2f_\mathrm{line}$ (100 or 120 Hz), which the loudspeaker reproduces as hum. The cause is that $V_{out}$ is measured against ground while $V_{CC}$ is not constant; the output is not insensitive to the supply. |
